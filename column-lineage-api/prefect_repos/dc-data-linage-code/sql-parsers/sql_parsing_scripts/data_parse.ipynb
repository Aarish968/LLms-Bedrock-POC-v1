```python
import os
import chardet
from typing import List, Dict, Generator


class InputLayer:
    """
    Input layer to collect and tokenize files from a codebase.
    Uses chardet for safe encoding detection.
    """
    def __init__(self, base_path: str, allowed_ext: List[str] = None, ignore_dirs: List[str] = None):
        self.base_path = base_path
        self.allowed_ext = allowed_ext or [".sql", ".py", ".ts", ".js"]
        self.ignore_dirs = set(ignore_dirs or ["__pycache__", ".git", "node_modules", "dist", "build", "venv"])

    def walk_directory(self) -> Generator[str, None, None]:
        """Traverse repo and yield files with allowed extensions, skipping ignored dirs."""
        for root, dirs, files in os.walk(self.base_path):
            # Modify dirs in-place → prevents os.walk from descending into ignored ones
            dirs[:] = [d for d in dirs if d not in self.ignore_dirs]

            for file in files:
                if not any(file.endswith(ext) for ext in self.allowed_ext):
                    continue
                yield os.path.join(root, file)

    def read_file(self, file_path: str) -> str:
        """Read file with encoding detection."""
        try:
            with open(file_path, "rb") as f:
                raw_data = f.read()
                encoding = chardet.detect(raw_data)["encoding"] or "utf-8"
            return raw_data.decode(encoding, errors="ignore")
        except Exception as e:
            print(f" Error reading {file_path}: {e}")
            return ""

    def tokenize_file(self, content: str, file_path: str) -> Dict:
        """Return file-level tokens (not line-based)."""
        return {
            "file_path": file_path,
            "content": content
        }

    def ingest(self) -> List[Dict]:
        """Collect all repo files as {path, content} dicts."""
        all_files = []
        for file_path in self.walk_directory():
            content = self.read_file(file_path)
            if content.strip():
                all_files.append(self.tokenize_file(content, file_path))
        return all_files


# Example run
if __name__ == "__main__":
    base_repo = "C:/dev/guided-workflow-backend"
  # replace with your repo path
    ingestor = InputLayer(base_repo, ignore_dirs=["__pycache__",".venv", ".git", "dist", "build",".env",".gitattributes",".gitignore","node_modules", "tests"])
    files = ingestor.ingest()

  

    print(f" Total files ingested: {len(files)}")
    print(" Sample:", files[0] if files else "No files found")

    


```

```python
import sqlglot
from sqlglot import expressions as exp
import re

class SQLParser:
    def __init__(self):
        pass

    def parse(self, files: list[dict]) -> dict:
        results = []
        
        print(f"Total files to check: {len(files)}")
        
        for i, f in enumerate(files):
            file_path = f["file_path"]
            
            # Only process .sql files for now
            if not file_path.lower().endswith('.sql'):
                continue
            
            print(f"Processing SQL file: {file_path}")
            
            try:
                # SUPER AGGRESSIVE CONTENT CLEANING
                content = f["content"]
                
                # Remove ALL ANSI escape codes
                content = re.sub(r'\x1b\[[0-9;]*[mGKH]', '', content)
                
                # Remove comment blocks with # symbols
                content = re.sub(r'^#{3,}.*?#{3,}$', '', content, flags=re.MULTILINE)
                content = re.sub(r'^#{10,}.*$', '', content, flags=re.MULTILINE)
                
                # Remove lines that are just # characters
                lines = content.split('\n')
                cleaned_lines = []
                for line in lines:
                    # Skip lines that are mostly # characters
                    if line.strip() and not re.match(r'^#{3,}', line.strip()):
                        cleaned_lines.append(line)
                
                content = '\n'.join(cleaned_lines)
                
                # Split by semicolons (proper SQL statement separation)
                statements = content.split(';')
                
                for stmt_content in statements:
                    stmt_content = stmt_content.strip()
                    
                    # Skip empty or very short statements
                    if not stmt_content or len(stmt_content) < 15:
                        continue
                    
                    # Skip comment-only statements
                    if stmt_content.startswith('--') or stmt_content.startswith('#'):
                        continue
                    
                    # Only process statements that look like complete SQL
                    if not any(stmt_content.upper().strip().startswith(keyword) 
                              for keyword in ['SELECT', 'INSERT', 'UPDATE', 'DELETE', 'CREATE', 'DROP', 'ALTER', 'WITH']):
                        continue
                    
                    try:
                        # Try parsing with Snowflake dialect
                        tree = sqlglot.parse(stmt_content, dialect="snowflake")
                        

                        for stmt in tree:
                            if stmt:

                                # print(stmt)
                                result = {
                                    "file": file_path,
                                    "type": self._safe_get_type(stmt),
                                    "tables": self._safe_get_tables(stmt),
                                    "columns": self._safe_get_columns(stmt),
                                    "joins": self._safe_get_joins(stmt),
                                    "sql_preview": stmt_content[:150]
                                }
                                results.append(result)
                                
                    except Exception as stmt_error:
                        # Only log significant errors
                        error_msg = str(stmt_error)
                        if len(stmt_content) > 30 and "Unexpected token" in error_msg:
                            results.append({
                                "file": file_path,
                                "error": f"Parse error: {error_msg[:100]}",
                                "statement_preview": stmt_content[:80]
                            })
                        
            except Exception as e:
                results.append({
                    "file": file_path,
                    "error": f"File error: {str(e)}"
                })
        
        successful = len([r for r in results if 'error' not in r])
        errors = len([r for r in results if 'error' in r])
        
        print(f"\nSuccessfully parsed statements: {successful}")
        print(f"Parsing errors: {errors}")
        
        return {"sql": results}
    
    def _safe_get_type(self, stmt):
        try:
            return stmt.key.upper() if hasattr(stmt, 'key') else "UNKNOWN"
        except:
            return "UNKNOWN"
    
    def _safe_get_tables(self, stmt):
        try:
            tables = []
            for t in stmt.find_all(exp.Table):
                try:
                    table_name = str(t.name) if hasattr(t, 'name') else str(t)
                    table_name = table_name.replace('"', '').replace("'", '')
                    if table_name and table_name != 'UNKNOWN':
                        tables.append(table_name)
                except:
                    pass
            return tables
        except:
            return []
    
    def _safe_get_columns(self, stmt):
        try:
            columns = []
            for c in stmt.find_all(exp.Column):
                try:
                    col_name = str(c.name) if hasattr(c, 'name') else str(c)
                    col_name = col_name.replace('"', '').replace("'", '')
                    if col_name and col_name != 'UNKNOWN':
                        columns.append(col_name)
                except:
                    pass
            return columns
        except:
            return []
    
    def _safe_get_joins(self, stmt):
        try:
            joins = []
            for j in stmt.find_all(exp.Join):
                try:
                    join_type = "INNER"
                    if hasattr(j, 'args') and 'kind' in j.args:
                        join_type = str(j.args.get("kind", "INNER")).upper()
                    
                    right_table = None
                    if hasattr(j, 'this') and j.this:
                        right_table = str(j.this.name) if hasattr(j.this, 'name') else str(j.this)
                    
                    if right_table:
                        joins.append({
                            "type": join_type,
                            "table": right_table
                        })
                except:
                    pass
            return joins
        except:
            return []

# Test the improved parser
if __name__ == "__main__":
    sample = files[:5]
    parser = SQLParser()
    result = parser.parse(sample)
    
    print("\n" + "="*60)
    print("FINAL RESULTS")
    print("="*60)
    
    successful = [r for r in result['sql'] if 'error' not in r]
    print(f"\nSuccessful parses: {len(successful)}")
    
    # Group by statement type
    by_type = {}
    for r in successful:
        stmt_type = r['type']
        if stmt_type not in by_type:
            by_type[stmt_type] = []
        by_type[stmt_type].append(r)
    
    print("\nStatement types found:")
    for stmt_type, statements in by_type.items():
        print(f"  {stmt_type}: {len(statements)} statements")
    import json
    
    print(json.dumps(by_type,indent =2))
    # Show sample of each type
    print("\nSample statements:")
    for stmt_type, statements in by_type.items():
        print(f"\n{stmt_type} Example:")
        r = statements[0]
        print(f"  Tables: {r['tables']}")
        print(f"  Columns: {r['columns'][:3]}...")
        if r['joins']:
            print(f"  Joins: {r['joins']}")
        print(f"  SQL: {r['sql_preview'][:100]}...")
    
    # Show remaining errors (should be much fewer)
    errors = [r for r in result['sql'] if 'error' in r]
    print(f"\nRemaining errors: {len(errors)}")
```

```python
import csv
import uuid
import json
from datetime import datetime
from typing import List, Dict, Optional

class ComprehensiveSQLToCSVGenerator:
    def __init__(self):
        self.entities = []
        self.relationships = []
        self.operations = []
        self.relationship_types = []
        self.entity_lookup = {}
        
        self._init_relationship_types()
    
    def _init_relationship_types(self):
        """Initialize comprehensive relationship types"""
        types = [
            ("CONTAINS_TABLE", "STRUCTURE", "Schema contains table"),
            ("CONTAINS_COLUMN", "STRUCTURE", "Table contains column"),
            ("REFERENCES_TABLE", "USAGE", "Statement references table"),
            ("JOINS_WITH", "USAGE", "Table joins with another table"),
            ("INSERTS_INTO", "EXECUTION", "Statement inserts into table"),
            ("SELECTS_FROM", "USAGE", "Statement selects from table"),
            ("CREATES_TABLE", "EXECUTION", "Statement creates table"),
            ("UPDATES_TABLE", "EXECUTION", "Statement updates table"),
            ("DELETES_FROM", "EXECUTION", "Statement deletes from table"),
            ("USES_COLUMN", "USAGE", "Statement uses column"),
            ("DEFINES_COLUMN", "STRUCTURE", "Table defines column"),
        ]
        
        for name, category, description in types:
            self.relationship_types.append({
                'relationship_type_id': str(uuid.uuid4()),
                'name': name,
                'category': category,
                'description': description
            })
    
    def process_sql_results(self, sql_results: Dict):
        """Comprehensively process SQL parsing results"""
        
        for result in sql_results['sql']:
            if 'error' in result:
                continue
            
            file_path = result['file']
            stmt_type = result['type']
            tables = result['tables']
            columns = result['columns']
            joins = result.get('joins', [])
            sql_preview = result.get('sql_preview', '')
            
            # Create file entity
            file_entity_id = self._add_entity(
                entity_name=file_path.split('\\')[-1],
                entity_type="SQL_FILE",
                qualified_name=file_path,
                description=f"SQL file containing {stmt_type} statements"
            )
            
            # Process each table
            table_entities = []
            for table_name in tables:
                # Handle schema.table format
                schema_name = None
                if '.' in table_name:
                    parts = table_name.split('.')
                    if len(parts) >= 2:
                        schema_name = '.'.join(parts[:-1])
                        table_simple_name = parts[-1]
                    else:
                        table_simple_name = table_name
                else:
                    table_simple_name = table_name
                
                # Create schema entity if exists
                schema_entity_id = None
                if schema_name:
                    schema_entity_id = self._add_entity(
                        entity_name=schema_name,
                        entity_type="SCHEMA",
                        qualified_name=schema_name,
                        description=f"Database schema"
                    )
                
                # Create table entity
                table_entity_id = self._add_entity(
                    entity_name=table_simple_name,
                    entity_type="TABLE",
                    qualified_name=table_name,
                    description=f"Table referenced in {stmt_type} statement",
                    parent_entity_id=schema_entity_id
                )
                table_entities.append(table_entity_id)
                
                # Create schema contains table relationship
                if schema_entity_id:
                    self._add_relationship(
                        source_entity_id=schema_entity_id,
                        target_entity_id=table_entity_id,
                        relationship_type="CONTAINS_TABLE",
                        detail=f"Schema contains table {table_simple_name}"
                    )
                
                # Create statement-table relationships based on type
                if stmt_type == "INSERT":
                    self._add_relationship(
                        source_entity_id=file_entity_id,
                        target_entity_id=table_entity_id,
                        relationship_type="INSERTS_INTO",
                        detail=f"INSERT statement targets table"
                    )
                elif stmt_type == "SELECT":
                    self._add_relationship(
                        source_entity_id=file_entity_id,
                        target_entity_id=table_entity_id,
                        relationship_type="SELECTS_FROM",
                        detail=f"SELECT statement reads from table"
                    )
                elif stmt_type == "CREATE":
                    self._add_relationship(
                        source_entity_id=file_entity_id,
                        target_entity_id=table_entity_id,
                        relationship_type="CREATES_TABLE",
                        detail=f"CREATE statement defines table"
                    )
                else:
                    self._add_relationship(
                        source_entity_id=file_entity_id,
                        target_entity_id=table_entity_id,
                        relationship_type="REFERENCES_TABLE",
                        detail=f"{stmt_type} statement references table"
                    )
            
            # Process columns
            for column_name in columns:
                # Try to associate column with table
                parent_table_id = table_entities[0] if table_entities else None
                
                # Handle table.column format
                if '.' in column_name:
                    parts = column_name.split('.')
                    if len(parts) >= 2:
                        table_part = parts[0]
                        col_part = parts[-1]
                        # Find matching table entity
                        for table_name in tables:
                            if table_part.lower() in table_name.lower():
                                parent_table_id = self.entity_lookup.get(table_name)
                                break
                        column_name = col_part
                
                # Create column entity
                column_qualified_name = f"{tables[0] if tables else 'unknown'}.{column_name}"
                column_entity_id = self._add_entity(
                    entity_name=column_name,
                    entity_type="COLUMN",
                    qualified_name=column_qualified_name,
                    description=f"Column used in {stmt_type} statement",
                    parent_entity_id=parent_table_id
                )
                
                # Create table contains column relationship
                if parent_table_id:
                    self._add_relationship(
                        source_entity_id=parent_table_id,
                        target_entity_id=column_entity_id,
                        relationship_type="CONTAINS_COLUMN",
                        detail=f"Table contains column {column_name}"
                    )
                
                # Create statement uses column relationship
                self._add_relationship(
                    source_entity_id=file_entity_id,
                    target_entity_id=column_entity_id,
                    relationship_type="USES_COLUMN",
                    detail=f"{stmt_type} statement uses column"
                )
            
            # Process joins
            for i, join in enumerate(joins):
                if i < len(table_entities) - 1:
                    self._add_relationship(
                        source_entity_id=table_entities[i],
                        target_entity_id=table_entities[i + 1],
                        relationship_type="JOINS_WITH",
                        detail=f"{join['type']} join between tables"
                    )
            
            # Create comprehensive operation record
            operation_metadata = {
                "statement_type": stmt_type,
                "tables_count": len(tables),
                "columns_count": len(columns),
                "joins_count": len(joins),
                "file_path": file_path
            }
            
            self._add_operation(
                operation_type=stmt_type,
                source_entity_id=file_entity_id,
                target_entity_id=table_entities[0] if table_entities else None,
                sql_snippet=sql_preview,
                execution_context=file_path,
                metadata=operation_metadata
            )
    
    def _add_entity(self, entity_name: str, entity_type: str, qualified_name: str, 
                   description: str = "", parent_entity_id: Optional[str] = None) -> str:
        """Add entity with comprehensive metadata"""
        if qualified_name in self.entity_lookup:
            return self.entity_lookup[qualified_name]
        
        entity_id = str(uuid.uuid4())
        entity = {
            'entity_id': entity_id,
            'entity_name': entity_name,
            'entity_type': entity_type,
            'qualified_name': qualified_name,
            'parent_entity_id': parent_entity_id or '',
            'description': description,
            'metadata': json.dumps({
                "entity_type": entity_type,
                "created_by": "SQL_PARSER",
                "source": "automated_analysis"
            }),
            'created_at': datetime.now().isoformat(),
            'updated_at': datetime.now().isoformat()
        }
        
        self.entities.append(entity)
        self.entity_lookup[qualified_name] = entity_id
        return entity_id
    
    def _add_relationship(self, source_entity_id: str, target_entity_id: str, 
                         relationship_type: str, detail: str = ""):
        """Add relationship with metadata"""
        rel_type_id = None
        for rt in self.relationship_types:
            if rt['name'] == relationship_type:
                rel_type_id = rt['relationship_type_id']
                break
        
        if not rel_type_id:
            return
        
        relationship = {
            'relationship_id': str(uuid.uuid4()),
            'source_entity_id': source_entity_id,
            'target_entity_id': target_entity_id,
            'relationship_type_id': rel_type_id,
            'relationship_detail': detail,
            'metadata': json.dumps({
                "relationship_type": relationship_type,
                "created_by": "SQL_PARSER"
            }),
            'created_at': datetime.now().isoformat()
        }
        self.relationships.append(relationship)
    
    def _add_operation(self, operation_type: str, source_entity_id: str, 
                      target_entity_id: Optional[str], sql_snippet: str,
                      execution_context: str, metadata: Dict):
        """Add comprehensive operation record"""
        operation = {
            'operation_id': str(uuid.uuid4()),
            'operation_type': operation_type,
            'source_entity_id': source_entity_id,
            'target_entity_id': target_entity_id or '',
            'performed_at': datetime.now().isoformat(),
            'performed_by': 'SQL_PARSER',
            'sql_snippet': sql_snippet,
            'execution_context': execution_context,
            'metadata': json.dumps(metadata)
        }
        self.operations.append(operation)
    
    def export_to_csv(self, output_dir: str = "comprehensive_csv"):
        """Export comprehensive data to CSV files"""
        from pathlib import Path
        
        output_path = Path(output_dir)
        output_path.mkdir(exist_ok=True)
        
        # Export entities
        with open(output_path / "entities.csv", 'w', newline='', encoding='utf-8') as f:
            fieldnames = ['entity_id', 'entity_name', 'entity_type', 'qualified_name',
                         'parent_entity_id', 'description', 'metadata', 'created_at', 'updated_at']
            writer = csv.DictWriter(f, fieldnames=fieldnames)
            writer.writeheader()
            writer.writerows(self.entities)
        
        # Export operations  
        with open(output_path / "operations.csv", 'w', newline='', encoding='utf-8') as f:
            fieldnames = ['operation_id', 'operation_type', 'source_entity_id', 'target_entity_id',
                         'performed_at', 'performed_by', 'sql_snippet', 'execution_context', 'metadata']
            writer = csv.DictWriter(f, fieldnames=fieldnames)
            writer.writeheader()
            writer.writerows(self.operations)
        
        # Export relationships
        with open(output_path / "relationships.csv", 'w', newline='', encoding='utf-8') as f:
            fieldnames = ['relationship_id', 'source_entity_id', 'target_entity_id',
                         'relationship_type_id', 'relationship_detail', 'metadata', 'created_at']
            writer = csv.DictWriter(f, fieldnames=fieldnames)
            writer.writeheader()
            writer.writerows(self.relationships)
        
        # Export relationship types
        with open(output_path / "relationship_types.csv", 'w', newline='', encoding='utf-8') as f:
            fieldnames = ['relationship_type_id', 'name', 'category', 'description']
            writer = csv.DictWriter(f, fieldnames=fieldnames)
            writer.writeheader()
            writer.writerows(self.relationship_types)
        
        print(f"\n✅ Comprehensive CSV files generated in '{output_dir}/':")
        print(f"   📄 entities.csv ({len(self.entities)} records)")
        print(f"   📄 operations.csv ({len(self.operations)} records)")
        print(f"   📄 relationships.csv ({len(self.relationships)} records)")
        print(f"   📄 relationship_types.csv ({len(self.relationship_types)} records)")
        
        # Print summary
        entity_types = {}
        for entity in self.entities:
            et = entity['entity_type']
            entity_types[et] = entity_types.get(et, 0) + 1
        
        print(f"\n📊 Entity Summary:")
        for entity_type, count in entity_types.items():
            print(f"   • {entity_type}: {count}")

# Use this comprehensive version
if __name__ == "__main__":
    # Run your SQL parser
    sample = files[:5]
    parser = SQLParser()
    sql_results = parser.parse(sample)
    print(sql_results[0])
    # Generate comprehensive CSV files
    # csv_generator = ComprehensiveSQLToCSVGenerator()
    # csv_generator.process_sql_results(sql_results)
    # csv_generator.export_to_csv("comprehensive_analysis")
```

```python
import csv
import uuid
import json
from datetime import datetime
from typing import List, Dict, Optional

class ComprehensiveSQLToCSVGenerator:
    def __init__(self):
        self.entities = []
        self.relationships = []
        self.operations = []
        self.relationship_types = []
        self.entity_lookup = {}
        self._init_relationship_types()
    
    def _init_relationship_types(self):
        """Initialize comprehensive relationship types"""
        types = [
            ("CONTAINS_TABLE", "STRUCTURE", "Schema contains table"),
            ("CONTAINS_COLUMN", "STRUCTURE", "Table contains column"),
            ("REFERENCES_TABLE", "USAGE", "Statement references table"),
            ("JOINS_WITH", "USAGE", "Table joins with another table"),
            ("INSERTS_INTO", "EXECUTION", "Statement inserts into table"),
            ("SELECTS_FROM", "USAGE", "Statement selects from table"),
            ("CREATES_TABLE", "EXECUTION", "Statement creates table"),
            ("UPDATES_TABLE", "EXECUTION", "Statement updates table"),
            ("DELETES_FROM", "EXECUTION", "Statement deletes from table"),
            ("USES_COLUMN", "USAGE", "Statement uses column"),
            ("DEFINES_COLUMN", "STRUCTURE", "Table defines column"),
            ("HAS_FILTER", "USAGE", "Statement has WHERE/HAVING filter"),
            ("HAS_AGGREGATE", "TRANSFORMATION", "Statement uses aggregate function"),
            ("HAS_WINDOW_FUNCTION", "TRANSFORMATION", "Statement uses window function"),
            ("HAS_SET_OPERATION", "TRANSFORMATION", "Statement uses UNION/INTERSECT/EXCEPT"),
        ]
        
        for name, category, description in types:
            self.relationship_types.append({
                'relationship_type_id': str(uuid.uuid4()),
                'name': name,
                'category': category,
                'description': description
            })
    
    def process_sql_results(self, sql_results: Dict):
        """Process SQL parsing results using ALL original functionalities"""
        
        for result in sql_results['sql']:
            if 'error' in result:
                continue
            
            file_path = result['file']
            stmt_type = result['type']
            tables = result.get('tables', [])
            columns = result.get('columns', [])
            joins = result.get('joins', [])
            filters = result.get('filters', [])
            grouping = result.get('grouping', [])
            aggregates = result.get('aggregates', [])
            window_functions = result.get('window_functions', [])
            set_ops = result.get('set_ops', [])
            expressions = result.get('expressions', [])
            ddl = result.get('ddl', None)
            
            # Create file entity
            file_entity_id = self._add_entity(
                entity_name=file_path.split('\\')[-1],
                entity_type="SQL_FILE",
                qualified_name=file_path,
                description=f"SQL file containing {stmt_type} statements"
            )
            
            # Process each table with full metadata
            table_entities = []
            for table_name in tables:
                # Handle schema.table format
                schema_name = None
                if '.' in table_name:
                    parts = table_name.split('.')
                    if len(parts) >= 2:
                        schema_name = '.'.join(parts[:-1])
                        table_simple_name = parts[-1]
                    else:
                        table_simple_name = table_name
                else:
                    table_simple_name = table_name
                
                # Create schema entity if exists
                schema_entity_id = None
                if schema_name:
                    schema_entity_id = self._add_entity(
                        entity_name=schema_name,
                        entity_type="SCHEMA",
                        qualified_name=schema_name,
                        description=f"Database schema"
                    )
                
                # Create table entity with rich metadata
                table_metadata = {
                    "statement_type": stmt_type,
                    "has_joins": len(joins) > 0,
                    "has_filters": len(filters) > 0,
                    "has_aggregates": len(aggregates) > 0,
                    "schema": schema_name
                }
                
                table_entity_id = self._add_entity(
                    entity_name=table_simple_name,
                    entity_type="TABLE",
                    qualified_name=table_name,
                    description=f"Table referenced in {stmt_type} statement",
                    parent_entity_id=schema_entity_id,
                    metadata=table_metadata
                )
                table_entities.append(table_entity_id)
                
                # Create schema contains table relationship
                if schema_entity_id:
                    self._add_relationship(
                        source_entity_id=schema_entity_id,
                        target_entity_id=table_entity_id,
                        relationship_type="CONTAINS_TABLE",
                        detail=f"Schema contains table {table_simple_name}"
                    )
                
                # Create statement-table relationships based on type
                if stmt_type == "INSERT":
                    self._add_relationship(
                        source_entity_id=file_entity_id,
                        target_entity_id=table_entity_id,
                        relationship_type="INSERTS_INTO",
                        detail=f"INSERT statement targets table"
                    )
                elif stmt_type == "SELECT":
                    self._add_relationship(
                        source_entity_id=file_entity_id,
                        target_entity_id=table_entity_id,
                        relationship_type="SELECTS_FROM",
                        detail=f"SELECT statement reads from table"
                    )
                elif stmt_type == "CREATE":
                    self._add_relationship(
                        source_entity_id=file_entity_id,
                        target_entity_id=table_entity_id,
                        relationship_type="CREATES_TABLE",
                        detail=f"CREATE statement defines table"
                    )
                else:
                    self._add_relationship(
                        source_entity_id=file_entity_id,
                        target_entity_id=table_entity_id,
                        relationship_type="REFERENCES_TABLE",
                        detail=f"{stmt_type} statement references table"
                    )
            
            # Process columns with full context
            for column_name in columns:
                # Try to associate column with table
                parent_table_id = table_entities[0] if table_entities else None
                
                # Handle table.column format
                if '.' in column_name:
                    parts = column_name.split('.')
                    if len(parts) >= 2:
                        table_part = parts[0]
                        col_part = parts[-1]
                        # Find matching table entity
                        for table_name in tables:
                            if table_part.lower() in table_name.lower():
                                parent_table_id = self.entity_lookup.get(table_name)
                                break
                        column_name = col_part
                
                # Create column entity with metadata
                column_qualified_name = f"{tables[0] if tables else 'unknown'}.{column_name}"
                column_metadata = {
                    "used_in_statement": stmt_type,
                    "used_in_filters": column_name in str(filters),
                    "used_in_grouping": column_name in str(grouping),
                    "used_in_aggregates": column_name in str(aggregates)
                }
                
                column_entity_id = self._add_entity(
                    entity_name=column_name,
                    entity_type="COLUMN",
                    qualified_name=column_qualified_name,
                    description=f"Column used in {stmt_type} statement",
                    parent_entity_id=parent_table_id,
                    metadata=column_metadata
                )
                
                # Create table contains column relationship
                if parent_table_id:
                    self._add_relationship(
                        source_entity_id=parent_table_id,
                        target_entity_id=column_entity_id,
                        relationship_type="CONTAINS_COLUMN",
                        detail=f"Table contains column {column_name}"
                    )
                
                # Create statement uses column relationship
                self._add_relationship(
                    source_entity_id=file_entity_id,
                    target_entity_id=column_entity_id,
                    relationship_type="USES_COLUMN",
                    detail=f"{stmt_type} statement uses column"
                )
            
            # Process joins with full detail from your original parser
            for join in joins:
                join_type = join.get('type', 'INNER')
                left_table = join.get('left')
                right_table = join.get('right')
                
                # Find entity IDs for join tables
                left_entity_id = None
                right_entity_id = None
                
                if left_table:
                    left_entity_id = self.entity_lookup.get(left_table)
                if right_table:
                    right_entity_id = self.entity_lookup.get(right_table)
                
                # Create join relationship
                if left_entity_id and right_entity_id:
                    self._add_relationship(
                        source_entity_id=left_entity_id,
                        target_entity_id=right_entity_id,
                        relationship_type="JOINS_WITH",
                        detail=f"{join_type} join between tables",
                        metadata={"join_type": join_type}
                    )
            
            # Process filters (WHERE/HAVING clauses)
            if filters:
                for filter_expr in filters:
                    self._add_relationship(
                        source_entity_id=file_entity_id,
                        target_entity_id=table_entities[0] if table_entities else file_entity_id,
                        relationship_type="HAS_FILTER",
                        detail=f"Statement has filter condition",
                        metadata={"filter_expression": str(filter_expr)[:200]}
                    )
            
            # Process aggregates
            if aggregates:
                for agg_func in aggregates:
                    self._add_relationship(
                        source_entity_id=file_entity_id,
                        target_entity_id=table_entities[0] if table_entities else file_entity_id,
                        relationship_type="HAS_AGGREGATE",
                        detail=f"Statement uses aggregate function",
                        metadata={"aggregate_function": str(agg_func)}
                    )
            
            # Process window functions
            if window_functions:
                for window_func in window_functions:
                    self._add_relationship(
                        source_entity_id=file_entity_id,
                        target_entity_id=table_entities[0] if table_entities else file_entity_id,
                        relationship_type="HAS_WINDOW_FUNCTION",
                        detail=f"Statement uses window function",
                        metadata={"window_function": str(window_func)}
                    )
            
            # Process set operations
            if set_ops:
                for set_op in set_ops:
                    self._add_relationship(
                        source_entity_id=file_entity_id,
                        target_entity_id=table_entities[0] if table_entities else file_entity_id,
                        relationship_type="HAS_SET_OPERATION",
                        detail=f"Statement uses set operation",
                        metadata={"set_operation": set_op}
                    )
            
            # Create comprehensive operation record with ALL your original data
            operation_metadata = {
                "statement_type": stmt_type,
                "tables_count": len(tables),
                "columns_count": len(columns),
                "joins_count": len(joins),
                "filters_count": len(filters),
                "aggregates_count": len(aggregates),
                "window_functions_count": len(window_functions),
                "set_ops_count": len(set_ops),
                "expressions_count": len(expressions),
                "has_ddl": ddl is not None,
                "file_path": file_path,
                "tables": tables,
                "columns": columns[:10],  # Limit for metadata size
                "joins": joins,
                "aggregates": aggregates,
                "window_functions": window_functions,
                "set_ops": set_ops
            }
            
            self._add_operation(
                operation_type=stmt_type,
                source_entity_id=file_entity_id,
                target_entity_id=table_entities[0] if table_entities else None,
                sql_snippet=str(result.get('sql_preview', ''))[:500],  # Longer snippet
                execution_context=file_path,
                metadata=operation_metadata
            )
    
    def _add_entity(self, entity_name: str, entity_type: str, qualified_name: str,
                   description: str = "", parent_entity_id: Optional[str] = None, 
                   metadata: Dict = None) -> str:
        """Add entity with comprehensive metadata"""
        if qualified_name in self.entity_lookup:
            return self.entity_lookup[qualified_name]
        
        entity_id = str(uuid.uuid4())
        
        # Merge default metadata with provided metadata
        default_metadata = {
            "entity_type": entity_type,
            "created_by": "SQL_PARSER",
            "source": "automated_analysis"
        }
        if metadata:
            default_metadata.update(metadata)
        
        entity = {
            'entity_id': entity_id,
            'entity_name': entity_name,
            'entity_type': entity_type,
            'qualified_name': qualified_name,
            'parent_entity_id': parent_entity_id or '',
            'description': description,
            'metadata': json.dumps(default_metadata),
            'created_at': datetime.now().isoformat(),
            'updated_at': datetime.now().isoformat()
        }
        
        self.entities.append(entity)
        self.entity_lookup[qualified_name] = entity_id
        return entity_id
    
    def _add_relationship(self, source_entity_id: str, target_entity_id: str,
                         relationship_type: str, detail: str = "", metadata: Dict = None):
        """Add relationship with metadata"""
        rel_type_id = None
        for rt in self.relationship_types:
            if rt['name'] == relationship_type:
                rel_type_id = rt['relationship_type_id']
                break
        
        if not rel_type_id:
            return
        
        # Merge default metadata with provided metadata
        default_metadata = {
            "relationship_type": relationship_type,
            "created_by": "SQL_PARSER"
        }
        if metadata:
            default_metadata.update(metadata)
        
        relationship = {
            'relationship_id': str(uuid.uuid4()),
            'source_entity_id': source_entity_id,
            'target_entity_id': target_entity_id,
            'relationship_type_id': rel_type_id,
            'relationship_detail': detail,
            'metadata': json.dumps(default_metadata),
            'created_at': datetime.now().isoformat()
        }
        self.relationships.append(relationship)
    
    def _add_operation(self, operation_type: str, source_entity_id: str,
                      target_entity_id: Optional[str], sql_snippet: str,
                      execution_context: str, metadata: Dict):
        """Add comprehensive operation record"""
        operation = {
            'operation_id': str(uuid.uuid4()),
            'operation_type': operation_type,
            'source_entity_id': source_entity_id,
            'target_entity_id': target_entity_id or '',
            'performed_at': datetime.now().isoformat(),
            'performed_by': 'SQL_PARSER',
            'sql_snippet': sql_snippet,
            'execution_context': execution_context,
            'metadata': json.dumps(metadata)
        }
        self.operations.append(operation)
    
    def export_to_csv(self, output_dir: str = "comprehensive_csv"):
        """Export comprehensive data to CSV files"""
        from pathlib import Path
        
        output_path = Path(output_dir)
        output_path.mkdir(exist_ok=True)
        
        # Export entities
        with open(output_path / "entities.csv", 'w', newline='', encoding='utf-8') as f:
            fieldnames = ['entity_id', 'entity_name', 'entity_type', 'qualified_name',
                         'parent_entity_id', 'description', 'metadata', 'created_at', 'updated_at']
            writer = csv.DictWriter(f, fieldnames=fieldnames)
            writer.writeheader()
            writer.writerows(self.entities)
        
        # Export operations  
        with open(output_path / "operations.csv", 'w', newline='', encoding='utf-8') as f:
            fieldnames = ['operation_id', 'operation_type', 'source_entity_id', 'target_entity_id',
                         'performed_at', 'performed_by', 'sql_snippet', 'execution_context', 'metadata']
            writer = csv.DictWriter(f, fieldnames=fieldnames)
            writer.writeheader()
            writer.writerows(self.operations)
        
        # Export relationships
        with open(output_path / "relationships.csv", 'w', newline='', encoding='utf-8') as f:
            fieldnames = ['relationship_id', 'source_entity_id', 'target_entity_id',
                         'relationship_type_id', 'relationship_detail', 'metadata', 'created_at']
            writer = csv.DictWriter(f, fieldnames=fieldnames)
            writer.writeheader()
            writer.writerows(self.relationships)
        
        # Export relationship types
        with open(output_path / "relationship_types.csv", 'w', newline='', encoding='utf-8') as f:
            fieldnames = ['relationship_type_id', 'name', 'category', 'description']
            writer = csv.DictWriter(f, fieldnames=fieldnames)
            writer.writeheader()
            writer.writerows(self.relationship_types)
        
        print(f"\n✅ Comprehensive CSV files generated in '{output_dir}/':")
        print(f"   📄 entities.csv ({len(self.entities)} records)")
        print(f"   📄 operations.csv ({len(self.operations)} records)")
        print(f"   📄 relationships.csv ({len(self.relationships)} records)")
        print(f"   📄 relationship_types.csv ({len(self.relationship_types)} records)")
        
        # Print detailed summary using all your original data
        entity_types = {}
        for entity in self.entities:
            et = entity['entity_type']
            entity_types[et] = entity_types.get(et, 0) + 1
        
        print(f"\n📊 Entity Summary:")
        for entity_type, count in entity_types.items():
            print(f"   • {entity_type}: {count}")
        
        # Show relationship summary
        rel_types = {}
        for rel in self.relationships:
            # Get relationship type name
            rel_type_name = "UNKNOWN"
            for rt in self.relationship_types:
                if rt['relationship_type_id'] == rel['relationship_type_id']:
                    rel_type_name = rt['name']
                    break
            rel_types[rel_type_name] = rel_types.get(rel_type_name, 0) + 1
        
        print(f"\n🔗 Relationship Summary:")
        for rel_type, count in rel_types.items():
            print(f"   • {rel_type}: {count}")
        
        # Show operation summary
        op_types = {}
        for op in self.operations:
            op_type = op['operation_type']
            op_types[op_type] = op_types.get(op_type, 0) + 1
        
        print(f"\n⚡ Operation Summary:")
        for op_type, count in op_types.items():
            print(f"   • {op_type}: {count}")

# Usage with specific file analysis
if __name__ == "__main__":
    # Specify the exact file you want to analyze
    target_file_path = "C:/dev/guided-workflow-backend\statements.sql"
    
    
    # Find the specific file in your files list
    target_file = None
    for file in files:
        if file["file_path"] == target_file_path:
            target_file = file
            break
    
    if target_file:
        sample = [target_file]  # Analyze just this one file
        print(f"✅ Found target file: {target_file_path}")
        print(f"📄 Content length: {len(target_file['content'])} characters")
        
        # Run your SQL parser with ALL original functionalities
        print(f"\n🔍 Running SQL analysis...")
        parser = SQLParser()
        sql_results = parser.parse(sample)
        
        # Generate comprehensive CSV files using ALL extracted data
        print(f"\n📊 Generating comprehensive CSV files...")
        csv_generator = ComprehensiveSQLToCSVGenerator()
        csv_generator.process_sql_results(sql_results)
        csv_generator.export_to_csv("statements_comprehensive_analysis")
        
        print(f"\n🎉 Complete analysis finished for {target_file_path}")
        
    else:
        print(f"❌ File not found: {target_file_path}")
        print("\n💡 Available SQL files:")
        sql_files = [f for f in files if f["file_path"].lower().endswith('.sql')]
        for i, f in enumerate(sql_files):
            print(f"  {i+1}. {f['file_path']}")
```

```python
import sqlglot
from sqlglot import expressions as exp
import re
import json


class SQLParser:
    def __init__(self):
        pass

    def parse(self, files: list[dict]) -> dict:
        results = []

        print(f"Total files to check: {len(files)}")

        for i, f in enumerate(files):
            file_path = f["file_path"]

            # Only process .sql files
            if not file_path.lower().endswith(".sql"):
                continue

            print(f"Processing SQL file: {file_path}")

            try:
                content = f["content"]

                # Remove ALL ANSI escape codes
                content = re.sub(r"\x1b\[[0-9;]*[mGKH]", "", content)

                # Remove comment blocks
                content = re.sub(r"^#{3,}.*?#{3,}$", "", content, flags=re.MULTILINE)
                content = re.sub(r"^#{10,}.*$", "", content, flags=re.MULTILINE)

                # Remove lines that are just #
                lines = content.split("\n")
                cleaned_lines = []
                for line in lines:
                    if line.strip() and not re.match(r"^#{3,}", line.strip()):
                        cleaned_lines.append(line)
                content = "\n".join(cleaned_lines)

                # Split by semicolon
                statements = content.split(";")

                for stmt_content in statements:
                    stmt_content = stmt_content.strip()

                    if not stmt_content or len(stmt_content) < 15:
                        continue
                    if stmt_content.startswith("--") or stmt_content.startswith("#"):
                        continue
                    if not any(
                        stmt_content.upper().strip().startswith(keyword)
                        for keyword in [
                            "SELECT",
                            "INSERT",
                            "UPDATE",
                            "DELETE",
                            "CREATE",
                            "DROP",
                            "ALTER",
                            "WITH",
                        ]
                    ):
                        continue

                    try:
                        trees = sqlglot.parse(stmt_content, dialect="snowflake")

                        for stmt in trees:
                            if stmt:
                                result = {
                                    "file": file_path,
                                    "type": self._safe_get_type(stmt),
                                    "tables": self._safe_get_tables(stmt),
                                    "columns": self._safe_get_columns(stmt),
                                    "joins": self._safe_get_joins(stmt),
                                    "join_columns": self._safe_get_join_columns(stmt),
                                    "select_columns": self._safe_get_select_columns(stmt),
                                    "where_columns": self._safe_get_where_columns(stmt),
                                    "group_columns": self._safe_get_group_columns(stmt),
                                    "having_columns": self._safe_get_having_columns(stmt),
                                    "sql_preview": stmt_content[:150],
                                }
                                results.append(result)

                    except Exception as stmt_error:
                        error_msg = str(stmt_error)
                        if len(stmt_content) > 30 and "Unexpected token" in error_msg:
                            results.append(
                                {
                                    "file": file_path,
                                    "error": f"Parse error: {error_msg[:100]}",
                                    "statement_preview": stmt_content[:80],
                                }
                            )

            except Exception as e:
                results.append({"file": file_path, "error": f"File error: {str(e)}"})

        successful = len([r for r in results if "error" not in r])
        errors = len([r for r in results if "error" in r])

        print(f"\nSuccessfully parsed statements: {successful}")
        print(f"Parsing errors: {errors}")

        return {"sql": results}

    # ------------------- SAFE HELPERS -------------------

    def _safe_get_type(self, stmt):
        try:
            return stmt.key.upper() if hasattr(stmt, "key") else "UNKNOWN"
        except:
            return "UNKNOWN"

    def _safe_get_tables(self, stmt):
        try:
            tables = []
            for t in stmt.find_all(exp.Table):
                table_name = str(getattr(t, "name", t)).replace('"', "").replace("'", "")
                if table_name and table_name != "UNKNOWN":
                    tables.append(table_name)
            return tables
        except:
            return []

    def _safe_get_columns(self, stmt):
        try:
            cols = []
            for c in stmt.find_all(exp.Column):
                col_name = str(getattr(c, "name", c)).replace('"', "").replace("'", "")
                if col_name and col_name != "UNKNOWN":
                    cols.append(col_name)
            return cols
        except:
            return []

    def _safe_get_joins(self, stmt):
        try:
            joins = []
            for j in stmt.find_all(exp.Join):
                join_type = str(j.args.get("kind", "INNER")).upper()
                right_table = (
                    str(j.this.name) if hasattr(j.this, "name") else str(j.this)
                )
                if right_table:
                    joins.append({"type": join_type, "table": right_table})
            return joins
        except:
            return []

    def _safe_get_join_columns(self, stmt):
        try:
            join_cols = []
            for j in stmt.find_all(exp.Join):
                if j.args.get("on"):
                    join_cols.extend([c.sql() for c in j.args["on"].find_all(exp.Column)])
            return join_cols
        except:
            return []

    def _safe_get_select_columns(self, stmt):
        try:
            return [c.sql() for c in getattr(stmt, "selects", [])]
        except:
            return []

    def _safe_get_where_columns(self, stmt):
        try:
            cols = []
            for w in stmt.find_all(exp.Where):
                cols.extend([c.sql() for c in w.find_all(exp.Column)])
            return cols
        except:
            return []

    def _safe_get_group_columns(self, stmt):
        try:
            cols = []
            for g in stmt.find_all(exp.Group):
                cols.extend([c.sql() for c in g.find_all(exp.Column)])
            return cols
        except:
            return []

    def _safe_get_having_columns(self, stmt):
        try:
            cols = []
            for h in stmt.find_all(exp.Having):
                cols.extend([c.sql() for c in h.find_all(exp.Column)])
            return cols
        except:
            return []


# ------------------- TEST -------------------
if __name__ == "__main__":
    sample = files[:5]  # assumes files is defined
    parser = SQLParser()
    result = parser.parse(sample)

    print("\n" + "=" * 60)
    print("FINAL RESULTS")
    print("=" * 60)

    successful = [r for r in result["sql"] if "error" not in r]
    print(f"\nSuccessful parses: {len(successful)}")

    by_type = {}
    for r in successful:
        stmt_type = r["type"]
        by_type.setdefault(stmt_type, []).append(r)

    print("\nStatement types found:")
    for stmt_type, statements in by_type.items():
        print(f"  {stmt_type}: {len(statements)} statements")

    print(json.dumps(by_type, indent=2))
```

```python
def extract_url_from_args(args_node, source_code):
    """
    Extract a string representation of the URL from the arguments node.
    Handles:
      - Template literals: `${V2_URL}/announcements`
      - String concatenation: V2_URL + "/announcements"
      - Direct string: "/static/path"
    """
    if args_node is None or len(args_node.children) == 0:
        return ""

    first_arg = args_node.children[0]

    # Template literal
    if first_arg.type == "template_string":
        return source_code[first_arg.start_byte:first_arg.end_byte].decode()

    # String literal
    elif first_arg.type == "string":
        return source_code[first_arg.start_byte:first_arg.end_byte].decode()

    # Binary expression (concatenation)
    elif first_arg.type == "binary_expression":
        return source_code[first_arg.start_byte:first_arg.end_byte].decode()

    # Fallback: just take raw text of first argument
    return source_code[first_arg.start_byte:first_arg.end_byte].decode()


```

```python
import os
import re
import json
import libcst as cst
from tree_sitter import Language, Parser
import tree_sitter_typescript as ts_typescript

# ----------------------------
# CONFIG
# ----------------------------
IGNORE_DIRS = {"__pycache__", ".venv", ".git", "dist", "build", "node_modules", "tests"}

# ----------------------------
# UTILS
# ----------------------------
def normalize_url(url: str):
    """Replace dynamic path params with {param} for fuzzy matching"""
    url = re.sub(r"\$\{[^\}]+\}", "{param}", url)
    url = re.sub(r"\{[^\}]+\}", "{param}", url)
    return url

# ----------------------------
# FRONTEND PARSER (Tree-sitter)
# ----------------------------
TS_LANGUAGE = Language(ts_typescript.language_typescript())
ts_parser = Parser(TS_LANGUAGE)


import re

def traverse_ts(node, source_code, results=None, parent_function=None):
    """
    Recursively traverse a TypeScript AST node to capture frontend action functions
    and client.* calls with full arguments. URL extraction/normalization is done
    later in post-processing.
    """
    if results is None:
        results = []

    current_function = parent_function

    # --------------------------
    # 1. Detect function declarations
    # --------------------------
    if node.type == "function_declaration":
        func_name_node = node.child_by_field_name("name")
        if func_name_node:
            current_function = source_code[func_name_node.start_byte:func_name_node.end_byte].decode()

    # --------------------------
    # 2. Detect arrow functions assigned to const/let
    # --------------------------
    elif node.type == "lexical_declaration":  # const foo = () => {...}
        for child in node.children:
            if child.type == "variable_declarator":
                id_node = child.child_by_field_name("name")
                value_node = child.child_by_field_name("value")
                if id_node and value_node and value_node.type in ("arrow_function", "function"):
                    current_function = source_code[id_node.start_byte:id_node.end_byte].decode()
                    break

    # --------------------------
    # 3. Detect client.* calls
    # --------------------------
    if node.type == "call_expression":
        func_node = node.child_by_field_name("function")
        if func_node and func_node.type == "member_expression":
            obj_node = func_node.child_by_field_name("object")
            prop_node = func_node.child_by_field_name("property")
            if obj_node and prop_node:
                obj_name = source_code[obj_node.start_byte:obj_node.end_byte].decode()
                prop_name = source_code[prop_node.start_byte:prop_node.end_byte].decode()

                if obj_name == "client" and current_function:
                    args_node = node.child_by_field_name("arguments")
                    args_text = ""
                    if args_node:
                        args_text = source_code[args_node.start_byte:args_node.end_byte].decode()

                    results.append({
                        "action_function": current_function,
                        "frontend_service": f"{obj_name}.{prop_name}",
                        "frontend_args": args_text,
                        "raw_args_node": args_node,  # keep for post-processing
                    })

    # --------------------------
    # 4. Recurse on children
    # --------------------------
    for child in node.children:
        traverse_ts(child, source_code, results, current_function)

    for call in results:
        if call['frontend_service'].split(".")[1] in ["get","post","put","delete"]:
            if call['raw_args_node']:
                # extract first string/template literal from args
                first_arg = call['raw_args_node'].children[0]
                url_text = source_code[first_arg.start_byte:first_arg.end_byte].decode()
                call['url'] = normalize_url(url_text)
            else:
                call['url'] = ""
        else:
            call['url'] = ""
        # Remove raw AST node before saving
        call.pop("raw_args_node", None)



    return results


# --------------------------
# Helper for post-processing URL normalization
# --------------------------
def normalize_url(url: str) -> str:
    """Replace template variables like ${id} or {id} with {param}"""
    if not url:
        return ""
    url = re.sub(r"\$\{[^}]+\}", "{param}", url)
    url = re.sub(r"\{[^}]+\}", "{param}", url)

    return url.rstrip("/")






def parse_ts_file(path):
    try:
        with open(path, "rb") as f:
            source_code = f.read()
        tree = ts_parser.parse(source_code)
        return traverse_ts(tree.root_node, source_code)
    except Exception:
        return []

def parse_frontend(root_dir: str):
    results = []
    for dirpath, dirnames, filenames in os.walk(root_dir):
        dirnames[:] = [d for d in dirnames if d not in IGNORE_DIRS]
        for file in filenames:
            if not file.endswith((".ts", ".tsx", ".js")):
                continue
            path = os.path.join(dirpath, file)
            results.extend(parse_ts_file(path))
    return results
```

```python

# ----------------------------
# BACKEND PARSER (LibCST)
# ----------------------------
class APIVisitor(cst.CSTVisitor):
    def __init__(self):
        self.endpoints = []

    def visit_FunctionDef(self, node:cst.FunctionDef):
        if node.decorators:
            for dec in node.decorators:
                dec_code = cst.Module([]).code_for_node(dec.decorator)
                if dec_code.startswith("app."):
                    method = dec_code.split('.')[1].split('(')[0].upper()
                    route = ""
                    if "(" in dec_code:
                        route = dec_code.split('(',1)[1].rstrip(')').strip('"\'')
                        route = normalize_url(route)
                    self.endpoints.append({
                        "function": node.name.value,
                        "method": method,
                        "route": route
                    })

def parse_backend_file(file_path):
    try:
        with open(file_path, encoding="utf8") as f:
            module = cst.parse_module(f.read())
        visitor = APIVisitor()
        module.visit(visitor)
        return visitor.endpoints
    except Exception:
        return []

def parse_backend(root_dir: str):
    backend_endpoints = []
    for dirpath, dirnames, files in os.walk(root_dir):
        dirnames[:] = [d for d in dirnames if d not in IGNORE_DIRS]
        for file in files:
            if file.endswith(".py"):
                backend_endpoints += parse_backend_file(os.path.join(dirpath, file))
    return backend_endpoints
```

```python

# ----------------------------
# HYBRID MAPPING
# ----------------------------
def map_api(frontend_calls, backend_endpoints):
    mapping = []
    for call in frontend_calls:
        for ep in backend_endpoints:
            # fuzzy match: method in function + normalized route
            if ep['method'] in call['frontend_service'].upper() and ep['route'] in call['frontend_args']:
                mapping.append({
                    "action_function": call['action_function'],
                    "frontend_service": call['frontend_service'],
                    "frontend_args": call['frontend_args'],
                    "backend_function": ep['function'],
                    "backend_route": ep['route'],
                    "backend_method": ep['method']
                })
    return mapping
```

```python
import sqlglot
from sqlglot import expressions as exp
import re
import os
import chardet
from typing import List, Dict, Optional, Set

class RobustSnowflakeParser:
    """
    Robust parser specifically designed for Snowflake stored procedures
    with proper handling of IDENTIFIER(), variables, and complex CTEs
    """
    
    def __init__(self):
        # Track variables declared in the procedure
        self.declared_variables = {}
        
    def parse(self, files: list[dict]) -> dict:
        """Main parsing method"""
        results = []
        
        print(f"Total files to check: {len(files)}")
        
        for i, f in enumerate(files):
            file_path = f["file_path"]
            
            if not file_path.lower().endswith('.sql'):
                continue
            
            print(f"Processing SQL file: {file_path}")
            
            try:
                content = f["content"]
                content = self._clean_content(content)
                content_type = self._classify_sql_content(content)
                print(f"  Content classified as: {content_type}")
                
                if content_type == "STORED_PROCEDURE":
                    procedure_results = self._parse_snowflake_procedure(file_path, content)
                    results.extend(procedure_results)
                else:
                    simple_results = self._parse_simple_statements(file_path, content)
                    results.extend(simple_results)
                        
            except Exception as e:
                results.append({
                    "file": file_path,
                    "error": f"File error: {str(e)}"
                })
        
        successful = len([r for r in results if 'error' not in r])
        errors = len([r for r in results if 'error' in r])
        
        print(f"\nSuccessfully parsed statements: {successful}")
        print(f"Parsing errors: {errors}")
        
        return {"sql": results}
    
    def _clean_content(self, content: str) -> str:
        """Clean content while preserving structure"""
        content = re.sub(r'\x1b\[[0-9;]*[mGKH]', '', content)
        content = re.sub(r'^#{3,}.*?#{3,}$', '', content, flags=re.MULTILINE)
        content = re.sub(r'^#{10,}.*$', '', content, flags=re.MULTILINE)
        
        lines = content.split('\n')
        cleaned_lines = []
        for line in lines:
            if line.strip() and not re.match(r'^#{3,}', line.strip()):
                cleaned_lines.append(line)
        
        return '\n'.join(cleaned_lines)
    
    def _classify_sql_content(self, content: str) -> str:
        """Classify SQL content type"""
        content_upper = content.upper()
        
        procedure_indicators = [
            r'CREATE\s+(?:OR\s+REPLACE\s+)?PROCEDURE',
            r'CREATE\s+(?:OR\s+REPLACE\s+)?FUNCTION', 
            r'BEGIN\s*$',
            r'DECLARE\s+\w+',
            r'END\s*;?\s*$'
        ]
        
        if any(re.search(pattern, content_upper, re.MULTILINE) for pattern in procedure_indicators):
            return "STORED_PROCEDURE"
        elif content.count(';') > 1:
            return "MULTI_STATEMENT"
        else:
            return "SIMPLE_STATEMENT"
    
    def _parse_snowflake_procedure(self, file_path: str, content: str) -> List[Dict]:
        """Parse Snowflake stored procedure with full context awareness"""
        results = []
        
        print(f"  Using Snowflake-specific parsing...")
        
        # Step 1: Extract procedure metadata
        procedure_name = self._extract_procedure_name(content)
        print(f"  Procedure name: {procedure_name}")
        
        # Step 2: Extract declared variables
        self.declared_variables = self._extract_declared_variables(content)
        print(f"  Found {len(self.declared_variables)} declared variables: {list(self.declared_variables.keys())}")
        
        # Step 3: Extract procedure body
        procedure_body = self._extract_procedure_body(content)
        print(f"  Procedure body length: {len(procedure_body)} characters")
        
        # Step 4: Extract statements with full context
        statements = self._extract_complete_statements(procedure_body)
        print(f"  Extracted {len(statements)} complete statements")
        
        # Step 5: Process each statement
        for i, stmt_info in enumerate(statements):
            print(f"    Processing statement {i+1}: {stmt_info['type']}")
            
            # Parse with context-aware method
            parsed_result = self._parse_statement_with_context(file_path, stmt_info, procedure_name)
            if parsed_result:
                results.append(parsed_result)
                
                # Show what we found
                tables = parsed_result.get('tables', [])
                columns = parsed_result.get('columns', [])
                print(f"      ✅ Found {len(tables)} tables, {len(columns)} columns")
                if tables:
                    print(f"         Tables: {tables[:5]}...")
        
        return results
    
    def _extract_procedure_name(self, content: str) -> str:
        """Extract procedure name"""
        match = re.search(r'CREATE\s+(?:OR\s+REPLACE\s+)?PROCEDURE\s+([^\s(]+)', content, re.IGNORECASE)
        return match.group(1) if match else "UNKNOWN_PROCEDURE"
    
    def _extract_declared_variables(self, content: str) -> Dict[str, str]:
        """Extract all DECLARE statements and their values"""
        variables = {}
        
        # Find all DECLARE statements
        declare_pattern = r'DECLARE\s+([A-Z_][A-Z0-9_]*)\s*:=\s*[\'"]([^\'"]+)[\'"]'
        matches = re.findall(declare_pattern, content, re.IGNORECASE)
        
        for var_name, var_value in matches:
            variables[var_name] = var_value
            
        print(f"    Declared variables: {variables}")
        return variables
    
    def _extract_procedure_body(self, content: str) -> str:
        """Extract procedure body between BEGIN and END"""
        begin_match = re.search(r'\bBEGIN\b', content, re.IGNORECASE)
        if not begin_match:
            return content
        
        end_match = re.search(r'\bEND\s*;?\s*\$\$', content, re.IGNORECASE)
        if not end_match:
            end_match = re.search(r'\bEND\s*;?\s*$', content, re.IGNORECASE | re.MULTILINE)
        
        if end_match:
            return content[begin_match.end():end_match.start()]
        else:
            return content[begin_match.end():]
    
    def _extract_complete_statements(self, procedure_body: str) -> List[Dict]:
        """Extract complete SQL statements with proper boundaries"""
        statements = []
        
        # Split content into logical blocks
        # Look for major SQL keywords that start new statements
        statement_starters = [
            'CREATE', 'SELECT', 'INSERT', 'UPDATE', 'DELETE', 'WITH', 'RETURN'
        ]
        
        lines = procedure_body.split('\n')
        current_statement = []
        current_type = None
        in_statement = False
        paren_depth = 0
        
        for line_num, line in enumerate(lines):
            line_stripped = line.strip()
            
            # Skip empty lines and comments
            if not line_stripped or line_stripped.startswith('--'):
                if in_statement:
                    current_statement.append(line)
                continue
            
            # Check for statement starters
            line_upper = line_stripped.upper()
            starts_new_statement = False
            
            for starter in statement_starters:
                if line_upper.startswith(starter):
                    starts_new_statement = True
                    new_type = starter
                    break
            
            # Handle parentheses depth for complex statements
            paren_depth += line.count('(') - line.count(')')
            
            if starts_new_statement and paren_depth <= 0 and not in_statement:
                # Start new statement
                if current_statement and current_type:
                    # Save previous statement
                    stmt_text = '\n'.join(current_statement).strip()
                    if len(stmt_text) > 20:
                        statements.append({
                            'statement': stmt_text,
                            'type': current_type,
                            'line_start': line_num - len(current_statement) + 1,
                            'line_end': line_num
                        })
                
                current_statement = [line]
                current_type = new_type
                in_statement = True
                
            elif in_statement:
                current_statement.append(line)
                
                # Check for statement end
                if (line_stripped.endswith(';') and paren_depth <= 0) or \
                   (line_upper.startswith('RETURN') and paren_depth <= 0):
                    
                    stmt_text = '\n'.join(current_statement).strip()
                    if len(stmt_text) > 20:
                        statements.append({
                            'statement': stmt_text,
                            'type': current_type,
                            'line_start': line_num - len(current_statement) + 1,
                            'line_end': line_num
                        })
                    
                    current_statement = []
                    current_type = None
                    in_statement = False
        
        # Handle final statement
        if current_statement and current_type:
            stmt_text = '\n'.join(current_statement).strip()
            if len(stmt_text) > 20:
                statements.append({
                    'statement': stmt_text,
                    'type': current_type,
                    'line_start': len(lines) - len(current_statement),
                    'line_end': len(lines)
                })
        
        return statements
    
    def _parse_statement_with_context(self, file_path: str, stmt_info: Dict, procedure_name: str) -> Optional[Dict]:
        """Parse individual statement with full context awareness"""
        
        statement = stmt_info['statement']
        stmt_type = stmt_info['type']
        
        # Try SQLGlot first for simple cases
        sqlglot_result = self._try_sqlglot_parse(statement)
        
        if sqlglot_result and len(sqlglot_result.get('tables', [])) > 0:
            # SQLGlot succeeded
            result = {
                "file": file_path,
                "type": sqlglot_result['type'],
                "tables": sqlglot_result['tables'],
                "columns": sqlglot_result['columns'],
                "joins": sqlglot_result['joins'],
                "sql_preview": statement[:200],
                "procedure_name": procedure_name,
                "statement_context": f"PROCEDURE_{stmt_type}",
                "parsing_method": "SQLGLOT_SUCCESS",
                "line_range": f"{stmt_info.get('line_start', 0)}-{stmt_info.get('line_end', 0)}"
            }
            return result
        
        # SQLGlot failed or incomplete, use comprehensive regex parsing
        regex_result = self._comprehensive_regex_parse(statement, stmt_type)
        
        result = {
            "file": file_path,
            "type": stmt_type,
            "tables": regex_result['tables'],
            "columns": regex_result['columns'],
            "joins": regex_result['joins'],
            "sql_preview": statement[:200],
            "procedure_name": procedure_name,
            "statement_context": f"PROCEDURE_{stmt_type}",
            "parsing_method": "COMPREHENSIVE_REGEX",
            "line_range": f"{stmt_info.get('line_start', 0)}-{stmt_info.get('line_end', 0)}",
            "variables_used": regex_result['variables_used']
        }
        
        return result
    
    def _try_sqlglot_parse(self, statement: str) -> Optional[Dict]:
        """Try parsing with SQLGlot"""
        try:
            tree = sqlglot.parse(statement, dialect="snowflake")
            
            for stmt in tree:
                if stmt:
                    return {
                        'type': self._safe_get_type(stmt),
                        'tables': self._safe_get_tables(stmt),
                        'columns': self._safe_get_columns(stmt),
                        'joins': self._safe_get_joins(stmt)
                    }
        except Exception:
            pass
        
        return None
    
    def _comprehensive_regex_parse(self, statement: str, stmt_type: str) -> Dict:
        """Comprehensive regex-based parsing for complex Snowflake statements"""
        
        tables = set()
        columns = set()
        joins = []
        variables_used = set()
        
        # 1. Extract tables from various contexts
        table_patterns = [
            # Standard FROM/JOIN patterns
            r'(?:FROM|JOIN)\s+([a-zA-Z_][a-zA-Z0-9_]*(?:\.[a-zA-Z_][a-zA-Z0-9_]*)*)',
            # CREATE TABLE patterns
            r'CREATE\s+(?:OR\s+REPLACE\s+)?(?:TRANSIENT\s+)?TABLE\s+([a-zA-Z_][a-zA-Z0-9_]*(?:\.[a-zA-Z_][a-zA-Z0-9_]*)*)',
            # INSERT INTO patterns
            r'INSERT\s+INTO\s+([a-zA-Z_][a-zA-Z0-9_]*(?:\.[a-zA-Z_][a-zA-Z0-9_]*)*)',
            # UPDATE patterns
            r'UPDATE\s+([a-zA-Z_][a-zA-Z0-9_]*(?:\.[a-zA-Z_][a-zA-Z0-9_]*)*)',
            # DELETE FROM patterns
            r'DELETE\s+FROM\s+([a-zA-Z_][a-zA-Z0-9_]*(?:\.[a-zA-Z_][a-zA-Z0-9_]*)*)',
        ]
        
        for pattern in table_patterns:
            matches = re.findall(pattern, statement, re.IGNORECASE)
            for match in matches:
                clean_table = match.strip().replace('"', '').replace("'", '')
                if clean_table and len(clean_table) < 100:  # Reasonable table name length
                    tables.add(clean_table)
        
        # 2. Handle Snowflake IDENTIFIER() syntax
        identifier_pattern = r'IDENTIFIER\s*\(\s*:([A-Z_][A-Z0-9_]*)\s*\)'
        identifier_matches = re.findall(identifier_pattern, statement, re.IGNORECASE)
        
        for var_name in identifier_matches:
            variables_used.add(var_name)
            # If we know the variable value, add it as a table
            if var_name in self.declared_variables:
                table_name = self.declared_variables[var_name]
                tables.add(table_name)
                print(f"      Resolved IDENTIFIER(:{var_name}) -> {table_name}")
        
        # 3. Extract columns from SELECT statements
        if stmt_type in ['SELECT', 'CREATE']:
            column_patterns = [
                # Simple column references
                r'SELECT\s+([^FROM]+)\s+FROM',
                # Column references in WHERE clauses
                r'WHERE\s+([a-zA-Z_][a-zA-Z0-9_]*)\s*[=<>!]',
                # Column references in GROUP BY
                r'GROUP\s+BY\s+([^ORDER\s]+)',
                # Column references in ORDER BY
                r'ORDER\s+BY\s+([^LIMIT\s]+)',
            ]
            
            for pattern in column_patterns:
                matches = re.findall(pattern, statement, re.IGNORECASE | re.DOTALL)
                for match in matches:
                    # Parse column list
                    if ',' in match:
                        cols = [col.strip() for col in match.split(',')]
                    else:
                        cols = [match.strip()]
                    
                    for col in cols[:10]:  # Limit to prevent noise
                        # Clean column name
                        col = re.sub(r'\s+as\s+\w+', '', col, flags=re.IGNORECASE)
                        col = re.sub(r'[()"\']', '', col).strip()
                        
                        # Skip functions and complex expressions
                        if (col and col != '*' and 
                            not re.search(r'(COUNT|SUM|AVG|MAX|MIN|CASE|WHEN)\s*\(', col, re.IGNORECASE) and
                            len(col) < 50 and
                            not col.upper() in ['FROM', 'WHERE', 'GROUP', 'ORDER', 'SELECT']):
                            columns.add(col)
        
        # 4. Extract JOIN information
        join_patterns = [
            (r'INNER\s+JOIN\s+([a-zA-Z_][a-zA-Z0-9_]*)', 'INNER'),
            (r'LEFT\s+JOIN\s+([a-zA-Z_][a-zA-Z0-9_]*)', 'LEFT'),
            (r'RIGHT\s+JOIN\s+([a-zA-Z_][a-zA-Z0-9_]*)', 'RIGHT'),
            (r'FULL\s+(?:OUTER\s+)?JOIN\s+([a-zA-Z_][a-zA-Z0-9_]*)', 'FULL_OUTER'),
            (r'(?:^|\s)JOIN\s+([a-zA-Z_][a-zA-Z0-9_]*)', 'INNER'),  # Default JOIN
        ]
        
        for pattern, join_type in join_patterns:
            matches = re.findall(pattern, statement, re.IGNORECASE)
            for match in matches:
                joins.append({
                    'type': join_type,
                    'table': match.strip()
                })
        
        return {
            'tables': list(tables),
            'columns': list(columns)[:20],  # Limit columns to prevent noise
            'joins': joins,
            'variables_used': list(variables_used)
        }
    
    def _parse_simple_statements(self, file_path: str, content: str) -> List[Dict]:
        """Parse simple SQL statements"""
        results = []
        
        statements = content.split(';')
        
        for stmt_content in statements:
            stmt_content = stmt_content.strip()
            
            if not stmt_content or len(stmt_content) < 15:
                continue
            
            if not any(stmt_content.upper().strip().startswith(keyword) 
                      for keyword in ['SELECT', 'INSERT', 'UPDATE', 'DELETE', 'CREATE', 'DROP', 'ALTER', 'WITH']):
                continue
            
            try:
                tree = sqlglot.parse(stmt_content, dialect="snowflake")
                
                for stmt in tree:
                    if stmt:
                        result = {
                            "file": file_path,
                            "type": self._safe_get_type(stmt),
                            "tables": self._safe_get_tables(stmt),
                            "columns": self._safe_get_columns(stmt),
                            "joins": self._safe_get_joins(stmt),
                            "sql_preview": stmt_content[:150],
                            "parsing_method": "SIMPLE_SQLGLOT"
                        }
                        results.append(result)
                        
            except Exception as stmt_error:
                error_msg = str(stmt_error)
                if len(stmt_content) > 30 and "Unexpected token" in error_msg:
                    results.append({
                        "file": file_path,
                        "error": f"Parse error: {error_msg[:100]}",
                        "statement_preview": stmt_content[:80]
                    })
        
        return results
    
    # Helper methods (keep existing ones)
    def _safe_get_type(self, stmt):
        try:
            return stmt.key.upper() if hasattr(stmt, 'key') else "UNKNOWN"
        except:
            return "UNKNOWN"
    
    def _safe_get_tables(self, stmt):
        try:
            tables = []
            for t in stmt.find_all(exp.Table):
                try:
                    table_name = str(t.name) if hasattr(t, 'name') else str(t)
                    table_name = table_name.replace('"', '').replace("'", '')
                    if table_name and table_name != 'UNKNOWN':
                        tables.append(table_name)
                except:
                    pass
            return tables
        except:
            return []
    
    def _safe_get_columns(self, stmt):
        try:
            columns = []
            for c in stmt.find_all(exp.Column):
                try:
                    col_name = str(c.name) if hasattr(c, 'name') else str(c)
                    col_name = col_name.replace('"', '').replace("'", '')
                    if col_name and col_name != 'UNKNOWN':
                        columns.append(col_name)
                except:
                    pass
            return columns
        except:
            return []
    
    def _safe_get_joins(self, stmt):
        try:
            joins = []
            for j in stmt.find_all(exp.Join):
                try:
                    join_type = "INNER"
                    if hasattr(j, 'args') and 'kind' in j.args:
                        join_type = str(j.args.get("kind", "INNER")).upper()
                    
                    right_table = None
                    if hasattr(j, 'this') and j.this:
                        right_table = str(j.this.name) if hasattr(j.this, 'name') else str(j.this)
                    
                    if right_table:
                        joins.append({
                            "type": join_type,
                            "table": right_table
                        })
                except:
                    pass
            return joins
        except:
            return []


def test_robust_parser():
    """Test the robust parser"""
    
    specific_file_path = r"C:\Users\nitallur\guided-workflow-backend\scripts\sql\Process_Revenue.sql"
    
    if not os.path.exists(specific_file_path):
        print(f"❌ File not found: {specific_file_path}")
        return
    
    try:
        with open(specific_file_path, "rb") as f:
            raw_data = f.read()
            encoding = chardet.detect(raw_data)["encoding"] or "utf-8"
        content = raw_data.decode(encoding, errors="ignore")
        
        print(f"✅ Successfully read file: {specific_file_path}")
        
    except Exception as e:
        print(f"❌ Error reading file: {e}")
        return
    
    test_files = [{
        "file_path": specific_file_path,
        "content": content
    }]
    
    print(f"\n🔍 Testing with ROBUST Snowflake Parser...")
    print("=" * 60)
    
    parser = RobustSnowflakeParser()
    result = parser.parse(test_files)
    
    print(f"\n📊 FINAL RESULTS:")
    print(f"Total statements found: {len(result['sql'])}")
    
    # Analyze results
    successful = [r for r in result['sql'] if 'error' not in r]
    print(f"✅ Successful parses: {len(successful)}")
    
    # Show detailed breakdown
    by_method = {}
    all_tables = set()
    all_columns = set()
    
    for r in successful:
        method = r.get('parsing_method', 'UNKNOWN')
        if method not in by_method:
            by_method[method] = []
        by_method[method].append(r)
        
        all_tables.update(r.get('tables', []))
        all_columns.update(r.get('columns', []))
    
    print(f"\n🔧 PARSING METHODS:")
    for method, statements in by_method.items():
        print(f"  • {method}: {len(statements)} statements")
    
    print(f"\n📋 COMPREHENSIVE RESULTS:")
    print("=" * 60)
    
    for i, stmt in enumerate(successful):
        print(f"\nStatement {i+1}:")
        print(f"  Type: {stmt.get('type', 'UNKNOWN')}")
        print(f"  Method: {stmt.get('parsing_method', 'UNKNOWN')}")
        print(f"  Tables ({len(stmt.get('tables', []))}): {stmt.get('tables', [])}")
        print(f"  Columns ({len(stmt.get('columns', []))}): {stmt.get('columns', [])[:5]}...")
        print(f"  Variables: {stmt.get('variables_used', [])}")
        print(f"  Lines: {stmt.get('line_range', 'N/A')}")
        print(f"  SQL: {stmt.get('sql_preview', '')[:100]}...")
    
    print(f"\n🎯 SUMMARY:")
    print(f"  • Total unique tables: {len(all_tables)}")
    print(f"  • Total unique columns: {len(all_columns)}")
    print(f"  • Tables found: {list(all_tables)}")
    
    return result


if __name__ == "__main__":
    print("🚀 ROBUST SNOWFLAKE PARSER TEST")
    print("=" * 60)
    
    result = test_robust_parser()
    
    if result and len(result['sql']) > 0:
        print(f"\n🎉 PARSING COMPLETE!")
        print(f"This version should extract much more complete information.")
    else:
        print(f"\n❌ Still having issues.")
```

```python
import sqlglot
from sqlglot import expressions as exp
import re
import os
import chardet
import csv
import json
from datetime import datetime
from typing import List, Dict, Optional, Set
from pathlib import Path

class RobustSnowflakeParser:
    """
    Robust parser specifically designed for Snowflake stored procedures
    with proper handling of IDENTIFIER(), variables, and complex CTEs
    """
    
    def __init__(self):
        # Track variables declared in the procedure
        self.declared_variables = {}
        
    def parse(self, files: list[dict]) -> dict:
        """Main parsing method"""
        results = []
        
        print(f"Total files to check: {len(files)}")
        
        for i, f in enumerate(files):
            file_path = f["file_path"]
            
            if not file_path.lower().endswith('.sql'):
                continue
            
            print(f"Processing SQL file: {file_path}")
            
            try:
                content = f["content"]
                content = self._clean_content(content)
                content_type = self._classify_sql_content(content)
                print(f"  Content classified as: {content_type}")
                
                if content_type == "STORED_PROCEDURE":
                    procedure_results = self._parse_snowflake_procedure(file_path, content)
                    results.extend(procedure_results)
                else:
                    simple_results = self._parse_simple_statements(file_path, content)
                    results.extend(simple_results)
                        
            except Exception as e:
                results.append({
                    "file": file_path,
                    "error": f"File error: {str(e)}"
                })
        
        successful = len([r for r in results if 'error' not in r])
        errors = len([r for r in results if 'error' in r])
        
        print(f"\nSuccessfully parsed statements: {successful}")
        print(f"Parsing errors: {errors}")
        
        return {"sql": results}
    
    def _clean_content(self, content: str) -> str:
        """Clean content while preserving structure"""
        content = re.sub(r'\x1b\[[0-9;]*[mGKH]', '', content)
        content = re.sub(r'^#{3,}.*?#{3,}$', '', content, flags=re.MULTILINE)
        content = re.sub(r'^#{10,}.*$', '', content, flags=re.MULTILINE)
        
        lines = content.split('\n')
        cleaned_lines = []
        for line in lines:
            if line.strip() and not re.match(r'^#{3,}', line.strip()):
                cleaned_lines.append(line)
        
        return '\n'.join(cleaned_lines)
    
    def _classify_sql_content(self, content: str) -> str:
        """Classify SQL content type"""
        content_upper = content.upper()
        
        procedure_indicators = [
            r'CREATE\s+(?:OR\s+REPLACE\s+)?PROCEDURE',
            r'CREATE\s+(?:OR\s+REPLACE\s+)?FUNCTION', 
            r'BEGIN\s*$',
            r'DECLARE\s+\w+',
            r'END\s*;?\s*$'
        ]
        
        if any(re.search(pattern, content_upper, re.MULTILINE) for pattern in procedure_indicators):
            return "STORED_PROCEDURE"
        elif content.count(';') > 1:
            return "MULTI_STATEMENT"
        else:
            return "SIMPLE_STATEMENT"
    
    def _parse_snowflake_procedure(self, file_path: str, content: str) -> List[Dict]:
        """Parse Snowflake stored procedure with full context awareness"""
        results = []
        
        print(f"  Using Snowflake-specific parsing...")
        
        # Step 1: Extract procedure metadata
        procedure_name = self._extract_procedure_name(content)
        print(f"  Procedure name: {procedure_name}")
        
        # Step 2: Extract declared variables
        self.declared_variables = self._extract_declared_variables(content)
        print(f"  Found {len(self.declared_variables)} declared variables: {list(self.declared_variables.keys())}")
        
        # Step 3: Extract procedure body
        procedure_body = self._extract_procedure_body(content)
        print(f"  Procedure body length: {len(procedure_body)} characters")
        
        # Step 4: Extract statements with full context
        statements = self._extract_complete_statements(procedure_body)
        print(f"  Extracted {len(statements)} complete statements")
        
        # Step 5: Process each statement
        for i, stmt_info in enumerate(statements):
            print(f"    Processing statement {i+1}: {stmt_info['type']}")
            
            # Parse with context-aware method
            parsed_result = self._parse_statement_with_context(file_path, stmt_info, procedure_name)
            if parsed_result:
                results.append(parsed_result)
                
                # Show what we found
                tables = parsed_result.get('tables', [])
                columns = parsed_result.get('columns', [])
                print(f"      ✅ Found {len(tables)} tables, {len(columns)} columns")
                if tables:
                    print(f"         Tables: {tables[:5]}...")
        
        return results
    
    def _extract_procedure_name(self, content: str) -> str:
        """Extract procedure name"""
        match = re.search(r'CREATE\s+(?:OR\s+REPLACE\s+)?PROCEDURE\s+([^\s(]+)', content, re.IGNORECASE)
        return match.group(1) if match else "UNKNOWN_PROCEDURE"
    
    def _extract_declared_variables(self, content: str) -> Dict[str, str]:
        """Extract all DECLARE statements and their values"""
        variables = {}
        
        # Find all DECLARE statements
        declare_pattern = r'DECLARE\s+([A-Z_][A-Z0-9_]*)\s*:=\s*[\'"]([^\'"]+)[\'"]'
        matches = re.findall(declare_pattern, content, re.IGNORECASE)
        
        for var_name, var_value in matches:
            variables[var_name] = var_value
            
        print(f"    Declared variables: {variables}")
        return variables
    
    def _extract_procedure_body(self, content: str) -> str:
        """Extract procedure body between BEGIN and END"""
        begin_match = re.search(r'\bBEGIN\b', content, re.IGNORECASE)
        if not begin_match:
            return content
        
        end_match = re.search(r'\bEND\s*;?\s*\$\$', content, re.IGNORECASE)
        if not end_match:
            end_match = re.search(r'\bEND\s*;?\s*$', content, re.IGNORECASE | re.MULTILINE)
        
        if end_match:
            return content[begin_match.end():end_match.start()]
        else:
            return content[begin_match.end():]
    
    def _extract_complete_statements(self, procedure_body: str) -> List[Dict]:
        """Extract complete SQL statements with proper boundaries"""
        statements = []
        
        # Split content into logical blocks
        statement_starters = [
            'CREATE', 'SELECT', 'INSERT', 'UPDATE', 'DELETE', 'WITH', 'RETURN'
        ]
        
        lines = procedure_body.split('\n')
        current_statement = []
        current_type = None
        in_statement = False
        paren_depth = 0
        
        for line_num, line in enumerate(lines):
            line_stripped = line.strip()
            
            # Skip empty lines and comments
            if not line_stripped or line_stripped.startswith('--'):
                if in_statement:
                    current_statement.append(line)
                continue
            
            # Check for statement starters
            line_upper = line_stripped.upper()
            starts_new_statement = False
            
            for starter in statement_starters:
                if line_upper.startswith(starter):
                    starts_new_statement = True
                    new_type = starter
                    break
            
            # Handle parentheses depth for complex statements
            paren_depth += line.count('(') - line.count(')')
            
            if starts_new_statement and paren_depth <= 0 and not in_statement:
                # Start new statement
                if current_statement and current_type:
                    # Save previous statement
                    stmt_text = '\n'.join(current_statement).strip()
                    if len(stmt_text) > 20:
                        statements.append({
                            'statement': stmt_text,
                            'type': current_type,
                            'line_start': line_num - len(current_statement) + 1,
                            'line_end': line_num
                        })
                
                current_statement = [line]
                current_type = new_type
                in_statement = True
                
            elif in_statement:
                current_statement.append(line)
                
                # Check for statement end
                if (line_stripped.endswith(';') and paren_depth <= 0) or \
                   (line_upper.startswith('RETURN') and paren_depth <= 0):
                    
                    stmt_text = '\n'.join(current_statement).strip()
                    if len(stmt_text) > 20:
                        statements.append({
                            'statement': stmt_text,
                            'type': current_type,
                            'line_start': line_num - len(current_statement) + 1,
                            'line_end': line_num
                        })
                    
                    current_statement = []
                    current_type = None
                    in_statement = False
        
        # Handle final statement
        if current_statement and current_type:
            stmt_text = '\n'.join(current_statement).strip()
            if len(stmt_text) > 20:
                statements.append({
                    'statement': stmt_text,
                    'type': current_type,
                    'line_start': len(lines) - len(current_statement),
                    'line_end': len(lines)
                })
        
        return statements
    
    def _parse_statement_with_context(self, file_path: str, stmt_info: Dict, procedure_name: str) -> Optional[Dict]:
        """Parse individual statement with full context awareness"""
        
        statement = stmt_info['statement']
        stmt_type = stmt_info['type']
        
        # Try SQLGlot first for simple cases
        sqlglot_result = self._try_sqlglot_parse(statement)
        
        if sqlglot_result and len(sqlglot_result.get('tables', [])) > 0:
            # SQLGlot succeeded
            result = {
                "file": file_path,
                "type": sqlglot_result['type'],
                "tables": sqlglot_result['tables'],
                "columns": sqlglot_result['columns'],
                "joins": sqlglot_result['joins'],
                "sql_preview": statement[:200],
                "procedure_name": procedure_name,
                "statement_context": f"PROCEDURE_{stmt_type}",
                "parsing_method": "SQLGLOT_SUCCESS",
                "line_range": f"{stmt_info.get('line_start', 0)}-{stmt_info.get('line_end', 0)}"
            }
            return result
        
        # SQLGlot failed or incomplete, use comprehensive regex parsing
        regex_result = self._comprehensive_regex_parse(statement, stmt_type)
        
        result = {
            "file": file_path,
            "type": stmt_type,
            "tables": regex_result['tables'],
            "columns": regex_result['columns'],
            "joins": regex_result['joins'],
            "sql_preview": statement[:200],
            "procedure_name": procedure_name,
            "statement_context": f"PROCEDURE_{stmt_type}",
            "parsing_method": "COMPREHENSIVE_REGEX",
            "line_range": f"{stmt_info.get('line_start', 0)}-{stmt_info.get('line_end', 0)}",
            "variables_used": regex_result['variables_used']
        }
        
        return result
    
    def _try_sqlglot_parse(self, statement: str) -> Optional[Dict]:
        """Try parsing with SQLGlot"""
        try:
            tree = sqlglot.parse(statement, dialect="snowflake")
            
            for stmt in tree:
                if stmt:
                    return {
                        'type': self._safe_get_type(stmt),
                        'tables': self._safe_get_tables(stmt),
                        'columns': self._safe_get_columns(stmt),
                        'joins': self._safe_get_joins(stmt)
                    }
        except Exception:
            pass
        
        return None
    
    def _comprehensive_regex_parse(self, statement: str, stmt_type: str) -> Dict:
        """Comprehensive regex-based parsing for complex Snowflake statements"""
        
        tables = set()
        columns = set()
        joins = []
        variables_used = set()
        
        # 1. Extract tables from various contexts
        table_patterns = [
            # Standard FROM/JOIN patterns
            r'(?:FROM|JOIN)\s+([a-zA-Z_][a-zA-Z0-9_]*(?:\.[a-zA-Z_][a-zA-Z0-9_]*)*)',
            # CREATE TABLE patterns
            r'CREATE\s+(?:OR\s+REPLACE\s+)?(?:TRANSIENT\s+)?TABLE\s+([a-zA-Z_][a-zA-Z0-9_]*(?:\.[a-zA-Z_][a-zA-Z0-9_]*)*)',
            # INSERT INTO patterns
            r'INSERT\s+INTO\s+([a-zA-Z_][a-zA-Z0-9_]*(?:\.[a-zA-Z_][a-zA-Z0-9_]*)*)',
            # UPDATE patterns
            r'UPDATE\s+([a-zA-Z_][a-zA-Z0-9_]*(?:\.[a-zA-Z_][a-zA-Z0-9_]*)*)',
            # DELETE FROM patterns
            r'DELETE\s+FROM\s+([a-zA-Z_][a-zA-Z0-9_]*(?:\.[a-zA-Z_][a-zA-Z0-9_]*)*)',
        ]
        
        for pattern in table_patterns:
            matches = re.findall(pattern, statement, re.IGNORECASE)
            for match in matches:
                clean_table = match.strip().replace('"', '').replace("'", '')
                if clean_table and len(clean_table) < 100:  # Reasonable table name length
                    tables.add(clean_table)
        
        # 2. Handle Snowflake IDENTIFIER() syntax
        identifier_pattern = r'IDENTIFIER\s*\(\s*:([A-Z_][A-Z0-9_]*)\s*\)'
        identifier_matches = re.findall(identifier_pattern, statement, re.IGNORECASE)
        
        for var_name in identifier_matches:
            variables_used.add(var_name)
            # If we know the variable value, add it as a table
            if var_name in self.declared_variables:
                table_name = self.declared_variables[var_name]
                tables.add(table_name)
                print(f"      Resolved IDENTIFIER(:{var_name}) -> {table_name}")
        
        # 3. Extract columns from SELECT statements
        if stmt_type in ['SELECT', 'CREATE']:
            column_patterns = [
                # Simple column references
                r'SELECT\s+([^FROM]+)\s+FROM',
                # Column references in WHERE clauses
                r'WHERE\s+([a-zA-Z_][a-zA-Z0-9_]*)\s*[=<>!]',
                # Column references in GROUP BY
                r'GROUP\s+BY\s+([^ORDER\s]+)',
                # Column references in ORDER BY
                r'ORDER\s+BY\s+([^LIMIT\s]+)',
            ]
            
            for pattern in column_patterns:
                matches = re.findall(pattern, statement, re.IGNORECASE | re.DOTALL)
                for match in matches:
                    # Parse column list
                    if ',' in match:
                        cols = [col.strip() for col in match.split(',')]
                    else:
                        cols = [match.strip()]
                    
                    for col in cols[:10]:  # Limit to prevent noise
                        # Clean column name
                        col = re.sub(r'\s+as\s+\w+', '', col, flags=re.IGNORECASE)
                        col = re.sub(r'[()"\']', '', col).strip()
                        
                        # Skip functions and complex expressions
                        if (col and col != '*' and 
                            not re.search(r'(COUNT|SUM|AVG|MAX|MIN|CASE|WHEN)\s*\(', col, re.IGNORECASE) and
                            len(col) < 50 and
                            not col.upper() in ['FROM', 'WHERE', 'GROUP', 'ORDER', 'SELECT']):
                            columns.add(col)
        
        # 4. Extract JOIN information
        join_patterns = [
            (r'INNER\s+JOIN\s+([a-zA-Z_][a-zA-Z0-9_]*)', 'INNER'),
            (r'LEFT\s+JOIN\s+([a-zA-Z_][a-zA-Z0-9_]*)', 'LEFT'),
            (r'RIGHT\s+JOIN\s+([a-zA-Z_][a-zA-Z0-9_]*)', 'RIGHT'),
            (r'FULL\s+(?:OUTER\s+)?JOIN\s+([a-zA-Z_][a-zA-Z0-9_]*)', 'FULL_OUTER'),
            (r'(?:^|\s)JOIN\s+([a-zA-Z_][a-zA-Z0-9_]*)', 'INNER'),  # Default JOIN
        ]
        
        for pattern, join_type in join_patterns:
            matches = re.findall(pattern, statement, re.IGNORECASE)
            for match in matches:
                joins.append({
                    'type': join_type,
                    'table': match.strip()
                })
        
        return {
            'tables': list(tables),
            'columns': list(columns)[:20],  # Limit columns to prevent noise
            'joins': joins,
            'variables_used': list(variables_used)
        }
    
    def _parse_simple_statements(self, file_path: str, content: str) -> List[Dict]:
        """Parse simple SQL statements"""
        results = []
        
        statements = content.split(';')
        
        for stmt_content in statements:
            stmt_content = stmt_content.strip()
            
            if not stmt_content or len(stmt_content) < 15:
                continue
            
            if not any(stmt_content.upper().strip().startswith(keyword) 
                      for keyword in ['SELECT', 'INSERT', 'UPDATE', 'DELETE', 'CREATE', 'DROP', 'ALTER', 'WITH']):
                continue
            
            try:
                tree = sqlglot.parse(stmt_content, dialect="snowflake")
                
                for stmt in tree:
                    if stmt:
                        result = {
                            "file": file_path,
                            "type": self._safe_get_type(stmt),
                            "tables": self._safe_get_tables(stmt),
                            "columns": self._safe_get_columns(stmt),
                            "joins": self._safe_get_joins(stmt),
                            "sql_preview": stmt_content[:150],
                            "parsing_method": "SIMPLE_SQLGLOT"
                        }
                        results.append(result)
                        
            except Exception as stmt_error:
                error_msg = str(stmt_error)
                if len(stmt_content) > 30 and "Unexpected token" in error_msg:
                    results.append({
                        "file": file_path,
                        "error": f"Parse error: {error_msg[:100]}",
                        "statement_preview": stmt_content[:80]
                    })
        
        return results
    
    # Helper methods
    def _safe_get_type(self, stmt):
        try:
            return stmt.key.upper() if hasattr(stmt, 'key') else "UNKNOWN"
        except:
            return "UNKNOWN"
    
    def _safe_get_tables(self, stmt):
        try:
            tables = []
            for t in stmt.find_all(exp.Table):
                try:
                    table_name = str(t.name) if hasattr(t, 'name') else str(t)
                    table_name = table_name.replace('"', '').replace("'", '')
                    if table_name and table_name != 'UNKNOWN':
                        tables.append(table_name)
                except:
                    pass
            return tables
        except:
            return []
    
    def _safe_get_columns(self, stmt):
        try:
            columns = []
            for c in stmt.find_all(exp.Column):
                try:
                    col_name = str(c.name) if hasattr(c, 'name') else str(c)
                    col_name = col_name.replace('"', '').replace("'", '')
                    if col_name and col_name != 'UNKNOWN':
                        columns.append(col_name)
                except:
                    pass
            return columns
        except:
            return []
    
    def _safe_get_joins(self, stmt):
        try:
            joins = []
            for j in stmt.find_all(exp.Join):
                try:
                    join_type = "INNER"
                    if hasattr(j, 'args') and 'kind' in j.args:
                        join_type = str(j.args.get("kind", "INNER")).upper()
                    
                    right_table = None
                    if hasattr(j, 'this') and j.this:
                        right_table = str(j.this.name) if hasattr(j.this, 'name') else str(j.this)
                    
                    if right_table:
                        joins.append({
                            "type": join_type,
                            "table": right_table
                        })
                except:
                    pass
            return joins
        except:
            return []


class ComprehensiveSQLToCSVGenerator:
    def __init__(self):
        self.entities = []
        self.relationships = []
        self.operations = []
        self.relationship_types = []
        self.entity_lookup = {}
        self._init_relationship_types()

    def _init_relationship_types(self):
        """Initialize comprehensive relationship types"""
        types = [
            ("CONTAINS_TABLE", "STRUCTURE", "Schema contains table"),
            ("CONTAINS_COLUMN", "STRUCTURE", "Table contains column"),
            ("REFERENCES_TABLE", "USAGE", "Statement references table"),
            ("JOINS_WITH", "USAGE", "Table joins with another table"),
            ("INSERTS_INTO", "EXECUTION", "Statement inserts into table"),
            ("SELECTS_FROM", "USAGE", "Statement selects from table"),
            ("CREATES_TABLE", "EXECUTION", "Statement creates table"),
            ("UPDATES_TABLE", "EXECUTION", "Statement updates table"),
            ("DELETES_FROM", "EXECUTION", "Statement deletes from table"),
            ("USES_COLUMN", "USAGE", "Statement uses column"),
            ("DEFINES_COLUMN", "STRUCTURE", "Table defines column"),
            ("USED_AS_SELECT", "USAGE", "Column selected in query"),
            ("USED_AS_JOIN", "USAGE", "Column used in JOIN condition"),
            ("USED_AS_FILTER", "USAGE", "Column used in WHERE clause"),
            ("USED_AS_GROUP_BY", "USAGE", "Column used in GROUP BY clause"),
            ("USED_AS_ORDER_BY", "USAGE", "Column used in ORDER BY clause"),
            ("USED_AS_HAVING", "USAGE", "Column used in HAVING clause"),
            ("COLUMN_UPDATED_BY", "USAGE", "Target column updated by source column"),
            ("USED_AS_UPDATE", "USAGE", "Column used in UPDATE clause"),
            ("PROCEDURE_CONTAINS_STATEMENT", "STRUCTURE", "Procedure contains SQL statement"),
            ("VARIABLE_RESOLVES_TO_TABLE", "USAGE", "Variable resolves to table name"),
        ]

        for name, category, description in types:
            self.relationship_types.append({
                'relationship_type_name': name,
                'relationship_category': category,
                'relationship_description': description
            })

    def process_sql_results(self, sql_results: Dict):
        """Enhanced processing for stored procedure results"""
        
        for result in sql_results['sql']:
            if 'error' in result:
                continue

            file_path = result['file']
            stmt_type = result['type']
            tables = result['tables']
            columns = result['columns']
            joins = result.get('joins', [])
            sql_preview = result.get('sql_preview', '')
            
            # Handle procedure-specific fields
            procedure_name = result.get('procedure_name')
            statement_context = result.get('statement_context', 'UNKNOWN')
            parsing_method = result.get('parsing_method', 'UNKNOWN')
            variables_used = result.get('variables_used', [])

            # Create file entity
            file_name = file_path.split('\\')[-1]
            self._add_entity(
                name=file_name,
                type="SQL_FILE",
                full_name=file_path,
                description=f"SQL file containing {stmt_type} statements"
            )

            # Create procedure entity if this is from a stored procedure
            if procedure_name and procedure_name != "UNKNOWN_PROCEDURE":
                self._add_entity(
                    name=procedure_name,
                    type="STORED_PROCEDURE",
                    full_name=f"{file_name}::{procedure_name}",
                    description=f"Stored procedure containing multiple SQL statements",
                    parent_name=file_name
                )

            # Create statement entity (for better granularity)
            statement_name = f"{procedure_name or file_name}_{stmt_type}_{len(self.entities)}"
            self._add_entity(
                name=statement_name,
                type="SQL_STATEMENT",
                full_name=f"{file_name}::{procedure_name or 'MAIN'}::{stmt_type}",
                description=f"{stmt_type} statement in {statement_context}",
                parent_name=procedure_name or file_name
            )

            # Process each table
            for table_name in tables:
                # Handle schema.table format
                schema_name = None
                if '.' in table_name:
                    parts = table_name.split('.')
                    if len(parts) >= 2:
                        schema_name = '.'.join(parts[:-1])
                        table_simple_name = parts[-1]
                    else:
                        table_simple_name = table_name
                else:
                    table_simple_name = table_name

                # Create schema entity if exists
                if schema_name:
                    self._add_entity(
                        name=schema_name,
                        type="SCHEMA",
                        full_name=schema_name,
                        description=f"Database schema"
                    )

                # Create table entity
                self._add_entity(
                    name=table_simple_name,
                    type="TABLE",
                    full_name=table_name,
                    description=f"Table referenced in {stmt_type} statement",
                    parent_name=schema_name
                )

                # Create schema contains table relationship
                if schema_name:
                    self._add_relationship(
                        source_name=schema_name,
                        target_name=table_simple_name,
                        relationship_type="CONTAINS_TABLE",
                        detail=f"Schema {schema_name} contains table {table_simple_name}"
                    )

                # Create statement-table relationships based on type
                if stmt_type == "INSERT":
                    self._add_relationship(
                        source_name=statement_name,
                        target_name=table_simple_name,
                        relationship_type="INSERTS_INTO",
                        detail=f"INSERT statement targets table {table_simple_name}"
                    )
                elif stmt_type == "SELECT":
                    self._add_relationship(
                        source_name=statement_name,
                        target_name=table_simple_name,
                        relationship_type="SELECTS_FROM",
                        detail=f"SELECT statement reads from table {table_simple_name}"
                    )
                elif stmt_type == "CREATE":
                    self._add_relationship(
                        source_name=statement_name,
                        target_name=table_simple_name,
                        relationship_type="CREATES_TABLE",
                        detail=f"CREATE statement defines table {table_simple_name}"
                    )
                else:
                    self._add_relationship(
                        source_name=statement_name,
                        target_name=table_simple_name,
                        relationship_type="REFERENCES_TABLE",
                        detail=f"{stmt_type} statement references table {table_simple_name}"
                    )

            # Process columns
            for column_name in columns:
                # Try to associate column with table
                parent_table_name = tables[0] if tables else None

                # Handle table.column format
                if '.' in column_name:
                    parts = column_name.split('.')
                    if len(parts) >= 2:
                        table_part = parts[0]
                        col_part = parts[-1]
                        # Find matching table
                        for table_name in tables:
                            if table_part.lower() in table_name.lower():
                                parent_table_name = table_name.split('.')[-1]  # Get simple table name
                                break
                        column_name = col_part

                # Create column entity
                column_full_name = f"{parent_table_name}.{column_name}" if parent_table_name else column_name
                self._add_entity(
                    name=column_name,
                    type="COLUMN",
                    full_name=column_full_name,
                    description=f"Column used in {stmt_type} statement",
                    parent_name=parent_table_name
                )

                # Create table contains column relationship
                if parent_table_name:
                    self._add_relationship(
                        source_name=parent_table_name,
                        target_name=column_name,
                        relationship_type="CONTAINS_COLUMN",
                        detail=f"Table {parent_table_name} contains column {column_name}"
                    )

                # Create statement uses column relationship
                                # Create statement uses column relationship
                self._add_relationship(
                    source_name=statement_name,
                    target_name=column_name,
                    relationship_type="USES_COLUMN",
                    detail=f"{stmt_type} statement uses column {column_name}"
                )

            # Process joins
            for i, join in enumerate(joins):
                if i < len(tables) - 1:
                    table1_name = tables[i].split('.')[-1]  # Get simple name
                    table2_name = tables[i + 1].split('.')[-1]  # Get simple name
                    self._add_relationship(
                        source_name=table1_name,
                        target_name=table2_name,
                        relationship_type="JOINS_WITH",
                        detail=f"{join['type']} join between {table1_name} and {table2_name}"
                    )

            # Process variables used (Snowflake specific)
            for var_name in variables_used:
                var_full_name = f"{procedure_name}.{var_name}" if procedure_name else var_name
                self._add_entity(
                    name=var_name,
                    type="PROCEDURE_VARIABLE",
                    full_name=var_full_name,
                    description=f"Variable used in stored procedure",
                    parent_name=procedure_name
                )

                # Create relationship between statement and variable
                self._add_relationship(
                    source_name=statement_name,
                    target_name=var_name,
                    relationship_type="REFERENCES_TABLE",
                    detail=f"Statement uses variable {var_name}"
                )

            # Create procedure contains statement relationship
            if procedure_name:
                self._add_relationship(
                    source_name=procedure_name,
                    target_name=statement_name,
                    relationship_type="PROCEDURE_CONTAINS_STATEMENT",
                    detail=f"Procedure {procedure_name} contains {stmt_type} statement"
                )

            # Create comprehensive operation record
            operation_metadata = {
                "statement_type": stmt_type,
                "tables_count": len(tables),
                "columns_count": len(columns),
                "joins_count": len(joins),
                "file_path": file_path,
                "procedure_name": procedure_name,
                "statement_context": statement_context,
                "parsing_method": parsing_method,
                "variables_used": variables_used,
                "line_range": result.get('line_range', 'N/A')
            }

            self._add_operation(
                operation_name=f"{statement_name}_operation",
                operation_type=stmt_type,
                source_name=statement_name,
                target_name=tables[0].split('.')[-1] if tables else None,
                sql_snippet=sql_preview,
                execution_context=f"{file_name}::{procedure_name or 'MAIN'}",
                metadata=operation_metadata
            )

    def _add_entity(self, name: str, type: str, full_name: str, 
                   description: str = "", parent_name: Optional[str] = None):
        """Add entity using actual names as identifiers"""
        # Use full_name as the unique identifier to avoid duplicates
        if full_name in self.entity_lookup:
            return  # Entity already exists
        
        entity = {
            'name': name,
            'type': type,
            'full_name': full_name,
            'parent_name': parent_name or '',
            'description': description,
            'additional_info': json.dumps({
                "entity_type": type,
                "created_by": "SQL_PARSER",
                "source": "automated_analysis"
            }),
            'created_date': datetime.now().isoformat(),
            'last_updated': datetime.now().isoformat()
        }

        self.entities.append(entity)
        self.entity_lookup[full_name] = name

    def _add_relationship(self, source_name: str, target_name: str, 
                         relationship_type: str, detail: str = ""):
        """Add relationship using actual names"""
        relationship = {
            'source_entity_name': source_name,
            'target_entity_name': target_name,
            'relationship_type': relationship_type,
            'relationship_details': detail,
            'additional_info': json.dumps({
                "relationship_type": relationship_type,
                "created_by": "SQL_PARSER"
            }),
            'created_date': datetime.now().isoformat()
        }
        self.relationships.append(relationship)

    def _add_operation(self, operation_name: str, operation_type: str, source_name: str, 
                      target_name: Optional[str], sql_snippet: str,
                      execution_context: str, metadata: Dict):
        """Add operation using actual names"""
        operation = {
            'operation_name': operation_name,
            'operation_type': operation_type,
            'source_entity_name': source_name,
            'target_entity_name': target_name or '',
            'executed_at': datetime.now().isoformat(),
            'executed_by': 'SQL_PARSER',
            'sql_code': sql_snippet,
            'execution_context': execution_context,
            'additional_details': json.dumps(metadata)
        }
        self.operations.append(operation)

    def export_to_csv(self, output_dir: str = "comprehensive_csv"):
        """Export comprehensive data to CSV files using actual names"""
        output_path = Path(output_dir)
        output_path.mkdir(exist_ok=True)

        # Export entities with actual names as identifiers
        with open(output_path / "data_entities.csv", 'w', newline='', encoding='utf-8') as f:
            fieldnames = [
                'name', 'type', 'full_name', 'parent_name', 
                'description', 'additional_info', 'created_date', 'last_updated'
            ]
            writer = csv.DictWriter(f, fieldnames=fieldnames)
            writer.writeheader()
            writer.writerows(self.entities)

        # Export operations with actual names
        with open(output_path / "sql_operations.csv", 'w', newline='', encoding='utf-8') as f:
            fieldnames = [
                'operation_name', 'operation_type', 'source_entity_name', 'target_entity_name',
                'executed_at', 'executed_by', 'sql_code', 'execution_context', 'additional_details'
            ]
            writer = csv.DictWriter(f, fieldnames=fieldnames)
            writer.writeheader()
            writer.writerows(self.operations)

        # Export relationships with actual names
        with open(output_path / "entity_relationships.csv", 'w', newline='', encoding='utf-8') as f:
            fieldnames = [
                'source_entity_name', 'target_entity_name', 'relationship_type',
                'relationship_details', 'additional_info', 'created_date'
            ]
            writer = csv.DictWriter(f, fieldnames=fieldnames)
            writer.writeheader()
            writer.writerows(self.relationships)

        # Export relationship types
        with open(output_path / "relationship_types.csv", 'w', newline='', encoding='utf-8') as f:
            fieldnames = [
                'relationship_type_name', 'relationship_category', 'relationship_description'
            ]
            writer = csv.DictWriter(f, fieldnames=fieldnames)
            writer.writeheader()
            writer.writerows(self.relationship_types)

        print(f"\n✅ CSV files with actual entity names generated in '{output_dir}/':")
        print(f"   📄 data_entities.csv ({len(self.entities)} records)")
        print(f"   📄 sql_operations.csv ({len(self.operations)} records)")
        print(f"   📄 entity_relationships.csv ({len(self.relationships)} records)")
        print(f"   📄 relationship_types.csv ({len(self.relationship_types)} records)")

        # Print enhanced summary
        entity_types = {}
        for entity in self.entities:
            et = entity['type']
            entity_types[et] = entity_types.get(et, 0) + 1

        print(f"\n📊 Entity Summary:")
        for entity_type, count in entity_types.items():
            print(f"   • {entity_type}: {count}")

        # Show sample entities
        print(f"\n📋 Sample Entities:")
        for entity_type in ['STORED_PROCEDURE', 'TABLE', 'COLUMN']:
            samples = [e for e in self.entities if e['type'] == entity_type][:3]
            if samples:
                print(f"   {entity_type}:")
                for sample in samples:
                    print(f"      • {sample['name']} ({sample['full_name']})")

        print(f"\n📋 CSV Column Descriptions:")
        print(f"   📄 data_entities.csv:")
        print(f"      • name: Entity name (table_name, column_name, procedure_name)")
        print(f"      • type: Entity type (TABLE, COLUMN, PROCEDURE, etc.)")
        print(f"      • full_name: Complete qualified name with schema/path")
        print(f"      • parent_name: Name of parent entity")
        print(f"      • description: Human-readable description")
        
        print(f"\n   📄 entity_relationships.csv:")
        print(f"      • source_entity_name: Name of source entity")
        print(f"      • target_entity_name: Name of target entity")
        print(f"      • relationship_type: Type of relationship")
        print(f"      • relationship_details: Description of the relationship")
        
        print(f"\n   📄 sql_operations.csv:")
        print(f"      • operation_name: Name of the operation")
        print(f"      • operation_type: Type of SQL operation (SELECT, INSERT, etc.)")
        print(f"      • source_entity_name: Entity performing the operation")
        print(f"      • target_entity_name: Entity being operated on")
        print(f"      • sql_code: Actual SQL code snippet")


def test_complete_integration():
    """Test the complete integration with name-based CSV export"""
    
    specific_file_path = r"C:\Users\nitallur\guided-workflow-backend\scripts\sql\Process_Revenue.sql"
    
    if not os.path.exists(specific_file_path):
        print(f"❌ File not found: {specific_file_path}")
        return
    
    try:
        with open(specific_file_path, "rb") as f:
            raw_data = f.read()
            encoding = chardet.detect(raw_data)["encoding"] or "utf-8"
        content = raw_data.decode(encoding, errors="ignore")
        
        print(f"✅ Successfully read file: {specific_file_path}")
        
    except Exception as e:
        print(f"❌ Error reading file: {e}")
        return
    
    # Create test files in the format expected by the parser
    test_files = [{
        "file_path": specific_file_path,
        "content": content
    }]
    
    print(f"\n🔍 INTEGRATION TEST WITH NAME-BASED IDENTIFIERS")
    print("=" * 70)
    
    # Step 1: Parse with robust parser
    print("Step 1: Parsing SQL with RobustSnowflakeParser...")
    parser = RobustSnowflakeParser()
    sql_results = parser.parse(test_files)
    
    print(f"\nParsing completed: {len(sql_results['sql'])} statements found")
    
    # Step 2: Generate CSV files with name-based identifiers
    print("\nStep 2: Generating CSV files with name-based identifiers...")
    csv_generator = ComprehensiveSQLToCSVGenerator()
    csv_generator.process_sql_results(sql_results)
    csv_generator.export_to_csv("name_based_data_lineage")
    
    print(f"\n🎉 INTEGRATION TEST COMPLETED!")
    print(f"Check the 'name_based_data_lineage' folder for CSV files.")
    
    return sql_results, csv_generator


# Main execution
if __name__ == "__main__":
    print("🚀 SNOWFLAKE PARSER + NAME-BASED CSV EXPORT")
    print("=" * 70)
    
    # Run the complete integration test
    sql_results, csv_generator = test_complete_integration()
    
    if sql_results and len(sql_results['sql']) > 0:
        print(f"\n📊 FINAL STATISTICS:")
        print(f"   • SQL statements parsed: {len(sql_results['sql'])}")
        print(f"   • Entities created: {len(csv_generator.entities)}")
        print(f"   • Relationships created: {len(csv_generator.relationships)}")
        print(f"   • Operations recorded: {len(csv_generator.operations)}")
        
        print(f"\n✅ SUCCESS! CSV files now use actual entity names as identifiers:")
        print(f"   📄 data_entities.csv - Uses table names, column names, procedure names")
        print(f"   📄 entity_relationships.csv - Shows relationships between named entities")
        print(f"   📄 sql_operations.csv - Operations on named entities")
        print(f"   📄 relationship_types.csv - Types of relationships")
        
        print(f"\n💡 Benefits of name-based approach:")
        print(f"   • Easy to read and understand")
        print(f"   • Can directly search for table/column names")
        print(f"   • No need to lookup UUIDs")
        print(f"   • Perfect for business users and analysts")
    else:
        print(f"\n❌ No results generated. Check the file path and content.")
```

```python
import re
import pandas as pd
from collections import defaultdict
import os

# You may need to install openpyxl and pandas if not already installed
# pip install pandas openpyxl

def extract_entities(sql):
    # Find all table/view/procedure names (IDENTIFIER(:NAME) or direct names)
    entity_pattern = re.compile(r"IDENTIFIER\s*\(\s*:(\w+)\s*\)|CREATE\s+(?:OR\s+REPLACE\s+)?(?:TABLE|VIEW|PROCEDURE|FUNCTION|TRANSIENT\s+TABLE)\s+([\w\.]+)", re.IGNORECASE)
    entities = set()
    for match in entity_pattern.finditer(sql):
        print(f"Regex match: {match.groups()}")
        if match.group(1):
            entities.add(match.group(1))
        elif match.group(2):
            entities.add(match.group(2))
    return list(entities)

def extract_actions(sql):
    # Find all actions (SELECT, INSERT, UPDATE, DELETE, CREATE, etc.)
    action_pattern = re.compile(r"\b(SELECT|INSERT|UPDATE|DELETE|CREATE|DROP|ALTER|MERGE|WITH)\b", re.IGNORECASE)
    return [(m.group(1).upper(), m.start()) for m in action_pattern.finditer(sql)]

def extract_columns(sql):
    # Find all column names in SELECT, JOIN, WHERE, etc.
    # This is a simple regex, not a full SQL parser
    column_pattern = re.compile(r"\b([A-Z_][A-Z0-9_]*\.[A-Z_][A-Z0-9_]*)\b", re.IGNORECASE)
    return set(m.group(1) for m in column_pattern.finditer(sql))

def extract_relationships(sql):
    # Find dependencies: which entity uses which
    rels = []
    rel_pattern = re.compile(r"\b(FROM|JOIN|INTO|UPDATE|TABLE|VIEW|PROCEDURE)\s+([\w\.\:]+)", re.IGNORECASE)
    lines = sql.splitlines()
    # Alias mapping: {alias: real_entity}
    alias_map = {}
    # CTEs: {cte_name: definition}
    cte_pattern = re.compile(r"(\w+)\s+AS\s*\(", re.IGNORECASE)
    for i, line in enumerate(lines):
        # Alias detection in FROM/JOIN
        alias_match = re.search(r"(FROM|JOIN)\s+([\w\.:]+)\s+(AS\s+)?([\w]+)", line, re.IGNORECASE)
        if alias_match:
            real_entity = alias_match.group(2)
            alias = alias_match.group(4)
            alias_map[alias] = real_entity
        # CTE detection
        cte_match = cte_pattern.search(line)
        if cte_match:
            cte_name = cte_match.group(1)
            alias_map[cte_name] = 'CTE'
    # Main relationship extraction
    rel_pattern = re.compile(r"\b(FROM|JOIN|INTO|UPDATE|TABLE|VIEW|PROCEDURE)\s+([\w\.\:]+)(?:\s+(?:AS\s+)?([\w]+))?", re.IGNORECASE)
    for m in rel_pattern.finditer(sql):
        # Find line number
        start = m.start()
        line_num = sql[:start].count('\n') + 1
        entity = m.group(2)
        alias = m.group(3) if m.group(3) else None
        # Determine entity type
        if entity in alias_map and alias_map[entity] == 'CTE':
            entity_type = 'CTE'
        elif entity.isupper() and '_' in entity:
            entity_type = 'table_or_view'
        elif entity.startswith('IDENTIFIER'):
            entity_type = 'variable'
        else:
            entity_type = 'unknown'
        rels.append({
            'action': m.group(1).upper(),
            'entity': entity,
            'alias': alias,
            'context': m.group(1).upper(),
            'entity_type': entity_type,
            'line_num': line_num
        })
    # JOIN column extraction
    join_pattern = re.compile(r"JOIN\s+([\w\.\:]+)\s+(?:AS\s+)?([\w]+)?[^\n]*ON\s+([^\n]+)", re.IGNORECASE)
    join_cols = []
    for m in join_pattern.finditer(sql):
        join_entity = m.group(1)
        join_alias = m.group(2)
        on_clause = m.group(3)
        # Find columns in ON clause (e.g., A.COL1 = B.COL2)
        col_matches = re.findall(r"([\w]+)\.([\w]+)", on_clause)
        if len(col_matches) >= 2:
            join_cols.append({
                'left_entity': col_matches[0][0],
                'left_col': col_matches[0][1],
                'right_entity': col_matches[1][0],
                'right_col': col_matches[1][1],
                'join_entity': join_entity,
                'join_alias': join_alias
            })
        return rels, alias_map, join_cols

def extract_lineage(relationships, alias_map, join_cols, sql):
    # Build a comprehensive lineage: action, entity, column, context
    lineage = []
    for rel in relationships:
        entity = rel['entity']
        action = rel['action']
        context = rel['context']
        entity_type = rel.get('entity_type', 'unknown')
        alias = rel.get('alias')
        line_num = rel.get('line_num')
        entity_clean = entity.replace('IDENTIFIER(:','').replace(')','').replace(':','')
        # Find columns for this entity
        col_pattern = re.compile(rf"\b{re.escape(entity_clean)}\.([A-Z_][A-Z0-9_]*)\b", re.IGNORECASE)
        columns = set(m.group(1) for m in col_pattern.finditer(sql))
        if columns:
            for col in columns:
                lineage.append({'Action': action, 'Entity': entity_clean, 'Alias': alias, 'Column': col, 'Context': context, 'EntityType': entity_type, 'LineNum': line_num, 'Parent': None})
        else:
            lineage.append({'Action': action, 'Entity': entity_clean, 'Alias': alias, 'Column': None, 'Context': context, 'EntityType': entity_type, 'LineNum': line_num, 'Parent': None})
    # Add JOIN column relationships
    for join in join_cols:
        lineage.append({
            'Action': 'JOIN',
            'Entity': join['join_entity'],
            'Alias': join['join_alias'],
            'Column': f"{join['left_col']}={join['right_col']}",
            'Context': 'JOIN_ON',
            'EntityType': 'join',
            'LineNum': None,
            'Parent': f"{join['left_entity']}={join['right_entity']}"
        })
    return lineage

def main(sql_file, excel_file):
    with open(sql_file, 'r', encoding='utf-8') as f:
        sql = f.read()
        print(f"Read SQL file: {sql}")

    relationships, alias_map, join_cols = extract_relationships(sql)
    lineage = extract_lineage(relationships, alias_map, join_cols, sql)
    # Prepare DataFrame for comprehensive lineage
    df_lineage = pd.DataFrame(lineage)
    # Write to Excel (single sheet)
    with pd.ExcelWriter(excel_file, engine='openpyxl') as writer:
        df_lineage.to_excel(writer, sheet_name='Lineage', index=False)
    print(f"Excel file created: {excel_file}")

if __name__ == "__main__":
    sql_file = r"C:\Users\nitallur\guided-workflow-backend\scripts\sql\Process_Revenue.sql"   # Change if needed
    excel_file = "sql_lineage.xlsx"
    main(sql_file, excel_file)

```
